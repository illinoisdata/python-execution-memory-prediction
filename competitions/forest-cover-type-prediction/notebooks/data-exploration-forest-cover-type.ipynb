# Hypothesis 1

***Is sunlight abundant in NW and SE directions , are they away from the mountains, how does hillshade influence the growth of certain species ?***

**Sunlight in NW and SE Directions**
SE Direction:

SE-facing slopes in Theodore Roosevelt National Park receive abundant morning sunlight, which is less intense than afternoon sunlight.

Morning sunlight warms these slopes without causing excessive heat, making them ideal for vegetation that thrives in moderate sunlight and moist environments. This explains the high concentration of Cottonwood/Willow trees in this direction, as these riparian species rely on consistent water availability from nearby river basins like the Little Missouri River.

NW Direction:

NW-facing slopes receive moderate sunlight, primarily during the late afternoon. These slopes are cooler due to reduced exposure to direct sunlight during the day.

The rugged terrain of the badlands creates natural shading on NW-facing slopes, especially in the morning. This shading helps retain soil moisture and reduces evaporation, favoring species like junipers that prefer shaded and cooler conditions.

Both NW and SE slopes are part of the rugged badlands terrain, characterized by steep valleys, buttes, and ridges formed by erosion.

SE Slopes: These are often associated with low-lying floodplains near river basins (e.g., Little Missouri River). While not "away" from the mountains, they are closer to water sources, which enhances soil moisture and supports riparian forests.

NW Slopes: These slopes are more shaded due to nearby ridges or buttes, but they are still part of the badlands' mountainous terrain.

**Hillshade's Influence on Tree Growth**
Hillshade plays a critical role in shaping microclimates by controlling sunlight exposure and soil moisture:

Shading Effects:

NW-facing slopes experience more shading due to their orientation and nearby landforms like buttes. This reduces evaporation rates and maintains cooler soil temperatures, creating favorable conditions for drought-resistant species like junipers.

SE-facing slopes receive less shading in the morning, allowing for better photosynthesis while avoiding the stress of intense afternoon heat.

**Soil Moisture Retention:**

Shaded areas (e.g., NW slopes) retain more soil moisture, which is crucial for vegetation survival in arid or semi-arid environments like the badlands.

Hillshade helps mitigate the effects of drought by reducing water loss through evaporation.

**Vegetation Distribution**

Hillshade influences where specific tree species grow by creating microhabitats with varying levels of sunlight and moisture.

For example:

Riparian species like Cottonwood/Willow dominate SE-facing floodplains with abundant morning light and water.

Coniferous species like junipers thrive on cooler NW-facing slopes with moderate sunlight and better moisture retention.

**Quick empirical test**

Using Covertype distance variables (fast):
Compare distance-to-hydrology(horizontal distance in meter to nearest surface water features) for Cottonwood/Willow points in the SE aspect sector vs other directions.
If SE Cottonwood points have significantly smaller horizontal distance to hydrology, that supports your river-basin explanation.

```python
# Define the bins and labels
bins = [ -22.5, 22.5, 67.5, 112.5, 157.5, 202.5, 247.5, 292.5, 337.5, 382.5 ]
labels = ['N','NE','E','SE','S','SW','W','NW','N']

# Create a polar plot
fig, ax = plt.subplots(subplot_kw={'projection': 'polar'}, figsize=(6,6))

# Convert bins to radians
bins_rad = np.radians(bins)

# Draw the sectors (pizza slices)
for i in range(len(labels)-1):
    theta1, theta2 = bins_rad[i], bins_rad[i+1]
    ax.fill_between(
        np.linspace(theta1, theta2, 100),
        0, 1, 
        alpha=0.2, 
        label=labels[i]
    )

# Add labels at the center of each sector
for i in range(len(labels)-1):
    theta_mid = np.radians((bins[i] + bins[i+1]) / 2)
    ax.text(theta_mid, 1.05, labels[i], ha='center', va='center', fontsize=12, fontweight='bold')

# Adjust polar plot so that 0° (North) is at the top and angle increases clockwise
ax.set_theta_zero_location("N")
ax.set_theta_direction(-1)

# Add degree labels: N=0°, E=90°, S=180°, W=270°
ax.set_xticks(np.radians([0, 90, 180, 270]))
ax.set_xticklabels(['0° (N)', '90° (E)', '180° (S)', '270° (W)'])

# Clean up radial axis
ax.set_yticklabels([])

ax.set_title("Compass Sectors with Degree Axes", fontsize=14, fontweight="bold")

plt.show()
```

```python
# define aspect sectors; SE roughly 112.5°–157.5° (center 135°)
#The function takes an angle in degrees and tells you which compass direction that angle points to.
def sector_from_aspect(aspect_deg):
    # returns 'SE' etc. Use 8 compass sectors (N, NE, E, SE, S, SW, W, NW)
    bins = [ -22.5, 22.5, 67.5, 112.5, 157.5, 202.5, 247.5, 292.5, 337.5, 382.5 ]
    labels = ['N','NE','E','SE','S','SW','W','NW','N']
    # map degrees into 0-360
    a = aspect_deg % 360
    import bisect
    idx = bisect.bisect_right(bins, a) - 1
    return labels[idx]

df['aspect_sector'] = df['Aspect'].apply(sector_from_aspect)

# select cottonwood/willow cover type (4)
cot = df[df['Cover_Type'] == 4]

# get horizontal distance distributions
dist_se = cot[cot['aspect_sector'] == 'SE']['Horizontal_Distance_To_Hydrology'].dropna()
dist_other = cot[cot['aspect_sector'] != 'SE']['Horizontal_Distance_To_Hydrology'].dropna()

print("n(SE)=", len(dist_se), "n(other)=", len(dist_other))
print("mean dist SE (m):", dist_se.mean(), "mean dist other (m):", dist_other.mean())

# simple statistical test (non-parametric if not normal)
tstat, pval = stats.ttest_ind(dist_se, dist_other, equal_var=False)
print("t-stat:", tstat, "p-value:", pval)
```

**Sample sizes:**

SE group: n=642

Other directions: n=1518

Mean distance to water:

SE group: 150.0 m

Other: 85.3 m

t-statistic: 10.24 (very large in magnitude)

p-value: 1.09 × 10⁻²³ (essentially zero)

The mean is the most common way to describe the “center” of a distribution.

If SE mean = 150 m and Other mean = 85 m, you immediately see SE sites are on average much farther from water.

The t-test itself is based on comparing means, so I used it to compare means.

**Interpretation**

* *Distance to water is significantly greater in the SE*

Cottonwood/Willow points in the SE are, on average, ~65 m farther from water than those in other directions.

That’s counterintuitive if we think of them as riparian (water-loving) species.

* *The difference is highly statistically significant*

A p-value that tiny (< 0.00000000000000000000001) means the probability this difference arose by chance is basically zero.

So this is a real pattern in your dataset, not noise.

Soil_Type3 = "Haploborolis – Rock outcrop complex, rubbly"

Let’s break it down:

1. Soil family: Haploborolis

Haploborolis soils are well-drained, coarse-textured soils.

Often shallow with minimal water retention in general, but rubbly areas can hold localized pockets of moisture in depressions.

These soils are common in mountainous or foothill areas.

2. Rock outcrop complex

Indicates that this soil often occurs with exposed bedrock.

Vegetation may grow in small soil pockets between rocks or in rubbly zones.

Such areas are typically less competitive for trees like conifers, allowing riparian or opportunistic species like Cottonwood/Willow to dominate.

3. Rubbly

The “rubbly” descriptor refers to broken rock fragments in the soil.

Benefits Cottonwood/Willow by:

Providing well-drained substrate (reduces root rot risk)

Keeping soil aerated

Allowing water to accumulate in small depressions, which is enough for seedlings.

4. Ecological interpretation

Even without rivers/floodplains nearby, these micro-sites provide sufficient water retention and reduced competition for Cottonwood/Willow.

SE slopes may expose more of this soil type because of aspect-driven erosion, sunlight, and slope conditions.

SE dominance holds even when controlling for soil type.

> If average hillshade at 9am (greater than 200 closer to 255)  is greater than hillshade at 3 pm (closer to 140) , what does it mean ?

In a typical 8-bit color representation where values range from 0 to 255, a value of 255 corresponds to the maximum intensity for a specific color channel. This means:

* For Red (R), a value of 255 represents full intensity red.
* For Green (G), a value of 255 represents full intensity green.
* For Blue (B), a value of 255 represents full intensity blue.
In the context of Hillshade or similar index values, a Hillshade index of 255 would indicate maximum illumination or brightness, meaning that location or point receives the maximum amount of sunlight or is fully exposed to light at the specific time of measurement.

Differences in elevation can lead to variations in shading due to the position of the sun in the sky and there are other factors that can contribute to differences in Hillshade values. To draw conclusions about the relation between elevation and shading, lets try to find their correlation.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming you have a DataFrame called 'df' with columns 'Elevation', 'Hillshade_9am', and 'Hillshade_3pm'

# Step 1: Extract elevation and hillshade values
elevation = df['Elevation']
hillshade_9am = df['Hillshade_9am']
hillshade_12pm = df['Hillshade_Noon']
hillshade_3pm = df['Hillshade_3pm']

# Step 2: Calculate correlation coefficients
correlation_9am = elevation.corr(hillshade_9am)
correlation_12pm = elevation.corr(hillshade_12pm)
correlation_3pm = elevation.corr(hillshade_3pm)

print("Correlation between Elevation and Hillshade at 9am:", correlation_9am)
print("Correlation between Elevation and Hillshade at Noon:", correlation_12pm)
print("Correlation between Elevation and Hillshade at 3pm:", correlation_3pm)
 
```

Correlation between Elevation and Hillshade at 9am: 0.0979
This correlation suggests a very weak positive linear relationship between elevation and hillshade at 9am. It indicates that higher elevations may have slightly higher levels of hillshade in the morning, but the correlation is very weak.

Correlation between Elevation and Hillshade at Noon: 0.2158
This correlation indicates a weak positive linear relationship between elevation and hillshade at noon. It suggests that higher elevations tend to have higher levels of hillshade around noon, implying that elevated areas receive more direct sunlight during this time.

Correlation between Elevation and Hillshade at 3pm: 0.0895
Similar to the correlation at 9am, this correlation also suggests a very weak positive linear relationship between elevation and hillshade at 3pm. It indicates that higher elevations may have slightly higher levels of hillshade in the afternoon, but the correlation is very weak.

These soil_types have number of 1 > 600 . which is good sample of data to distinguish between soiltypes for different forest covers.

```python
# Create an empty dictionary to store count data for each soil type
soil_type_data = {}

# Iterate over each soil type column
for i in range(1, 41):
    soil_type_column = f'Soil_Type{i}'
    soil_type_count = df[soil_type_column].sum()
    
    # Check if the count is greater than or equal to 600
    if soil_type_count >= 600:
        # Store the count data for each soil type column
        Soil_Type_count_1 = df[df[soil_type_column] == 1].groupby('Forest_Cover_Type_Names')[soil_type_column].count()
        soil_type_data[soil_type_column] = Soil_Type_count_1
 
plt.figure(figsize=(15, 10))

# Iterate over soil type data dictionary
for i, (soil_type_column, counts) in enumerate(soil_type_data.items(), 1):
    plt.subplot(4, 4, i)
    plt.bar(counts.index, counts.values, color='aquamarine')
    plt.xlabel('Forest Cover Type')
    plt.ylabel('Count of 1s')
    plt.title(f'{soil_type_column}')
    plt.xticks(rotation=45, ha='right')
    
    # Highlighting max value
    max_value_index = counts.values.argmax()  # Get index of max value
    plt.bar(counts.index[max_value_index], counts.values[max_value_index], color='teal')  # Highlight max value with red color

plt.tight_layout()
plt.show()
```
