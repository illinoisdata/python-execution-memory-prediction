```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

別々に処理したデータをくっつけたい

```python
train_df = pd.read_csv("../input/house-prices-advanced-regression-techniques/train.csv")
test_df = pd.read_csv("../input/house-prices-advanced-regression-techniques/test.csv")
sample_df = pd.read_csv("../input/house-prices-advanced-regression-techniques/sample_submission.csv")
```

## 数値変数

```python
numerical_columns = test_df.select_dtypes(include=[np.number]).columns
```

```python
numerical_columns
```

```python
X_num = train_df[numerical_columns].fillna(0)
X_num_test = test_df[numerical_columns].fillna(0)
```

```python
X_cat = train_df[["Id", "MSZoning"]]
# ここで対数を取る
y = np.log1p(train_df["SalePrice"])
X_cat_test = test_df[["Id", "MSZoning"]]
```

```python
X_cat_2 = pd.get_dummies(X_cat)
X_cat_test_2 = pd.get_dummies(X_cat_test)
```

```python
X_cat_2
```

２つのデータをくっつける

```python
X = pd.merge(X_num, X_cat_2)
X_test = pd.merge(X_num_test, X_cat_test_2)
```

```python
X
```

```python
X_test
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X, y, random_state=0)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
pred_train = model.predict(X_train)
pred_valid = model.predict(X_valid)
pred_test = model.predict(X_test)
```

```python
print(np.sqrt(mean_squared_error(y_train, pred_train)), np.sqrt(mean_squared_error(y_valid, pred_valid)))
```

```python
sub_df = pd.DataFrame({
    "Id": sample_df["Id"], 
    "SalePrice": np.expm1(pred_test)  # logを元に戻す
})
```

```python

sub_df.to_csv("simple_combine_log.csv", index=False)
```
