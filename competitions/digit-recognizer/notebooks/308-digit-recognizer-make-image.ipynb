## Import Libs

```python
from PIL import Image
import os
import pandas as pd
import numpy as np
```

## Make Image

```python
df_train = pd.read_csv('/kaggle/input/digit-recognizer/train.csv')
df_train = np.array(df_train.iloc[:,1:])
save_dirs = '/kaggle/working/image_train'
if not os.path.exists(save_dirs):
    os.makedirs(save_dirs)
for i in range(len(df_train)):
    img = df_train[i].reshape(28,28)
    img = Image.fromarray(np.uint8(img))
    img.save(os.path.join(save_dirs,f'{i}.jpg'))
```
