```python
###START OF LINEAR REGRESSION CODE
import torch, pandas as pd

df = pd.read_csv('/kaggle/input/salary-dataset/Salary_dataset.csv')
device = 'cuda' if torch.cuda.is_available() else 'cpu'

X = torch.tensor(df["YearsExperience"].values, dtype=torch.float32).view(-1, 1).to(device)
y = torch.tensor(df["Salary"].values, dtype=torch.float32).view(-1, 1).to(device)
y /= 10000

X = torch.hstack((torch.ones(len(X), 1), X))
theta = torch.randn((2, 1), device=device)
eta, epoch = 0.01, 15000

for i in range(epoch):
    theta -= eta * 2 / len(X) * X.T @ (X @ theta - y)

import matplotlib.pyplot as plt

linX = torch.linspace(0, 11, 2)
linX = linX.view(len(linX), 1)
plt.plot(linX, linX @ theta[1] + theta[0], label='Line of best fit', color='darkorange', alpha=0.9)
plt.scatter(X[:, 1], y, label='Datapoints', alpha=0.7)
plt.xlabel('X (in years of experience)')
plt.ylabel('Y (in 10 thousand dollars)')
plt.grid(linestyle='-.')
plt.legend()
plt.show()

print(f'MSE of our linear regression model: {torch.mean((X @ theta - y) ** 2)}')
```

```python
df = pd.read_csv("/kaggle/input/digit-recognizer/train.csv")

device = 'cuda' if torch.cuda.is_available() else 'cpu'
oldX = torch.tensor(df.drop("label", axis=1).values, dtype=torch.float32, device=device)
oldY = torch.tensor(df["label"].values, dtype=torch.long, device=device)

mask = (oldY == 0) | (oldY == 1)
X, Y = oldX[mask], oldY[mask]
split = int(len(X) * 6 / 7)
trainX = X[:split].float().to(device)
testX = X[split:].float().to(device)
trainY = Y[:split].float().view(split, 1).to(device)
testY = Y[split:].float().view(len(X) - split, 1).to(device)
trainX /= 255
testX /= 255

weight = torch.randn((len(trainX[0]), 1), device=device, requires_grad=True)
bias = torch.ones((1, 1), device=device, requires_grad=True)
eta, epoch = 0.1, 15000
criterion = torch.nn.BCEWithLogitsLoss()

for i in range(epoch):
    loss = criterion(trainX @ weight + bias, trainY)
    loss.backward()
    with torch.no_grad():
        weight -= eta * weight.grad
        bias -= eta * bias.grad
    weight.grad.zero_()
    bias.grad.zero_()

correct = 0
for i in range(len(testX)):
    correct += torch.round(torch.sigmoid(testX[i] @ weight + bias)).item() == testY[i].item()

print(f'Percent correct on test data: {correct / len(testX)}%')
print(f'BCE logit loss of our binary OCR model via sigmoid regression: {criterion(testX @ weight + bias, testY)}')
```
