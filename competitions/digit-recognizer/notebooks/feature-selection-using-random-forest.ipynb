```python
import seaborn as sns
```

```python
df = pd.read_csv('/kaggle/input/digit-recognizer/train.csv')
df.head()
```

```python
# Extracting rows and columns :-

X = df.iloc[:,1:]
y = df.iloc[:,0]
```

```python
sns.heatmap(X.iloc[3].values.reshape(28,28))

# so,it shows the heatmap of that digit at a particualr index.
```

```python
# import Random classifier and train the model on the whole data.

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()
rf.fit(X,y)
```

```python
# it tells the "feature importances" of each columns:-

rf.feature_importances_
```

```python
rf.feature_importances_.shape
```

```python
sns.heatmap(rf.feature_importances_.reshape(28,28))

# it shows that pixels at the 'centre of the image' is more important to classify a digit,than the pixels around the 'periphery of the image' 
# darker the areas shows less importance.
```
