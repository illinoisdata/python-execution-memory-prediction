# Digit Recognizer using RandomForestClassifier

### Importing libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

### Forming Directories

```python
base_dir = '/kaggle/input/digit-recognizer'
print(os.listdir(base_dir))

test_dir = os.path.join(base_dir, 'test.csv')
train_dir = os.path.join(base_dir, 'train.csv')
print("Training - ", train_dir)
print("Test - ", test_dir)
```

### Reading the .csv and formatting

```python
train_data = pd.read_csv(train_dir)
train_df = pd.DataFrame(train_data)
sample_submission = pd.read_csv('/kaggle/input/digit-recognizer/sample_submission.csv')

test_data = pd.read_csv(test_dir)
test_df = pd.DataFrame(test_data)

train_df.head()
```

```python
train_images = train_df.drop('label', axis=1).values
train_labels = train_df['label'].values

print(f"Shape of train_images = {train_images.shape}")
```

```python
test_images = test_df
```

#### Showing a sample image and it's pixel values

```python
image_ = train_images[5].reshape(28,28)

plt.imshow(image_ , cmap='gray')
plt.title("Sample Image")
plt.show()
```

```python
print(train_images[5])
```

### Normalizing the values

```python
train_images = train_images / 255.0
test_images = test_images / 255.0
```

### Showing sample images

```python
figs, axes = plt.subplots(5, 5, figsize=(18,18))
idx = 0
for row in range(5):
    for col in range(5):
        axes[row,col].imshow(train_images[idx].reshape(28,28), cmap='gray')
        axes[row,col].axis('off')
        idx += 1

plt.tight_layout()
plt.show()
```

### Peforming train and validation split


```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(train_images, train_labels,test_size=0.2, random_state=42)
```

### Creating the **RandomForestClassifier** model

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=125, random_state=42)
rf_model.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score

val_preds = rf_model.predict(X_val)
accuracy = accuracy_score(y_val, val_preds)
print(f"Validation Accuracy: {accuracy*100:.4f} %")
```

```python
preds = rf_model.predict(test_images.values)
```

```python
submission = pd.DataFrame({
    "ImageId": np.arange(1, len(preds)+1),
    "Label": preds
})
submission_path = "/kaggle/working/submission.csv"
submission.to_csv(submission_path, index=False)
print("Saved:", submission_path)
submission.head()
```
