```python
import pandas as pd
train_data = pd.read_csv('/kaggle/input/titanic/train.csv')
```

```python
train_data.info()
```

```python
from pycaret.classification import *

# Setup
s = setup(train, target = 'Survived', session_id = 123)
```

```python
# Compare Models
best_model = compare_models()
```

```python
print(best_model)
```

```python
# Analyze Model
evaluate_model(best_model)
```

```python
# Predictions
predict_model(best_model)
```

```python
# 1. 최종 모델 생성 (전체 train 데이터 사용)
final_model = finalize_model(best_model)

# 2. 모델 정보 확인
print("최종 모델:")
print(final_model)
```

```python
test_data = pd.read_csv('/kaggle/input/titanic/test.csv')
test_data.info()
```

```python
# 1. Test 데이터에 대한 예측
# predict_model()은 setup()에서 학습한 전처리를 자동 적용
predictions = predict_model(final_model, data=test_data)

# 2. 예측 결과 확인
print("예측 결과 샘플:")
print(predictions[['PassengerId', 'prediction_label']].head(10))

# 3. Submission 파일 생성
submission = pd.DataFrame({
    'PassengerId': test_data['PassengerId'],
    'Survived': predictions['prediction_label']
})

# 4. 파일 저장
submission.to_csv('submission.csv', index=False)

# 5. 생성된 파일 확인
print("\nSubmission 파일 생성 완료!")
print(f"제출할 예측 수: {len(submission)}")
```
