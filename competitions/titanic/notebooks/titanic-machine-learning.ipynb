```python
survived_sex_distribution = df1[df1['Survived'] == 1]['Sex'].value_counts()
not_survived_sex_distribution = df1[df1['Survived'] == 0]['Sex'].value_counts()
labels_survived = survived_sex_distribution.index
sizes_survived = survived_sex_distribution.values
labels_not_survived = not_survived_sex_distribution.index
sizes_not_survived = not_survived_sex_distribution.values

colors = ['#ff9999', '#66b3ff']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 7))

ax1.pie(sizes_survived, labels=labels_survived, autopct='%1.1f%%', colors=colors, startangle=140)
ax1.set_title('Sex Distribution of Survivors')

colors = ['#66b3ff','#ff9999']

ax2.pie(sizes_not_survived, labels=labels_not_survived, autopct='%1.1f%%', colors=colors, startangle=140)
ax2.set_title('Sex Distribution of Non-Survivors')

plt.show()
plt.show()
```

```python
df2=pd.read_csv('/kaggle/input/titanic/test.csv')
```

```python
df2.isnull().sum()
```

```python
df=pd.concat([df1,df2]) # iki veri setini birleştirdik
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Fare']=df['Fare'].fillna(df['Fare'].mean()) # Boş verileri ortalama ile doldurduk
```

```python
df['Embarked'].value_counts()
```

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
del df['Cabin'] # Cabin sütununu sildik
```

```python
df['Title']=df['Name'].str.extract(' ([A-Za-z]+\.)',expand=False)
```

```python
df['Title'].value_counts()
```

```python
df['Title']= df['Title'].replace(['Rev.','Dr.','Col.','Major.','Sir.','Capt.','Jonkheer.','Don.'],'Mr.')
```

```python
df['Title'].value_counts()
```

```python
df['Title']= df['Title'].replace(['Lady.','Dona.','Countess.','Mme.'],'Mrs.')
```

```python
df['Title'].value_counts()
```

```python
df['Title']= df['Title'].replace(['Ms.','Mlle.'],'Miss.')
```

```python
df['Title'].value_counts()
```

```python
df['Age']=df['Age'].fillna(df.groupby('Title')['Age'].transform('mean')) # Gruplayarak doldurduk
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df=df.drop(['Name','Ticket','PassengerId'],axis=1) #işimize yaramayan sütunları sildik
```

```python
df=pd.get_dummies(df,drop_first=True)
```

```python
df.head()
```
