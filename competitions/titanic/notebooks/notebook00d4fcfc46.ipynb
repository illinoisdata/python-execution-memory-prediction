```python
# 🛳️ Titanic - Machine Learning from Disaster
# Complete end-to-end code — ready to copy & paste into Kaggle Notebook

# Step 1: Import libraries
import numpy as np
import pandas as pd
import os
from sklearn.ensemble import RandomForestClassifier

# Step 2: Check input files
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

# Step 3: Load the training and test data
train_data = pd.read_csv("/kaggle/input/titanic/train.csv")
test_data = pd.read_csv("/kaggle/input/titanic/test.csv")

# Display first few rows of training data
print("\nTraining Data Preview:")
print(train_data.head())

# Step 4: Explore survival rate by gender
women = train_data.loc[train_data.Sex == 'female']["Survived"]
men = train_data.loc[train_data.Sex == 'male']["Survived"]

rate_women = sum(women) / len(women)
rate_men = sum(men) / len(men)

print("\n✅ Survival Rate Insights:")
print("% of women who survived:", round(rate_women * 100, 2))
print("% of men who survived:", round(rate_men * 100, 2))
print("\nObservation: Gender is a strong survival indicator — more women survived than men.")

# Step 5: Prepare features and target variable
y = train_data["Survived"]
features = ["Pclass", "Sex", "SibSp", "Parch"]

# Convert categorical features to numeric using one-hot encoding
X = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])

# Step 6: Build and train the Random Forest model
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X, y)

# Step 7: Generate predictions
predictions = model.predict(X_test)

# Step 8: Create submission file
output = pd.DataFrame({
    'PassengerId': test_data.PassengerId,
    'Survived': predictions
})
output.to_csv('submission.csv', index=False)
print("\n✅ Submission file 'submission.csv' successfully created!")

# Step 9: Display confirmation message
print("\n🎉 Titanic model completed successfully!")
```
