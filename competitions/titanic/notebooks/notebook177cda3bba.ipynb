```python
# 🛳️ Titanic - Machine Learning from Disaster
# Complete Notebook (Ready to Run Anywhere)
# --------------------------------------------------------
# This project predicts which passengers survived the Titanic disaster
# using a Random Forest Classifier. It also creates a submission.csv file
# formatted for Kaggle competition submission.

# ✅ Step 1: Import Required Libraries
import numpy as np
import pandas as pd
import os
from sklearn.ensemble import RandomForestClassifier

# ✅ Step 2: Check if Dataset Exists
train_path = "train.csv"
test_path = "test.csv"

if not (os.path.exists(train_path) and os.path.exists(test_path)):
    print("⚠️ Titanic dataset not found! Creating a sample dataset for demonstration...")
    
    # Create a sample Titanic-like dataset
    train_data = pd.DataFrame({
        'PassengerId': [1, 2, 3, 4, 5, 6],
        'Survived': [0, 1, 1, 1, 0, 0],
        'Pclass': [3, 1, 3, 1, 3, 2],
        'Sex': ['male', 'female', 'female', 'female', 'male', 'male'],
        'SibSp': [1, 1, 0, 1, 0, 0],
        'Parch': [0, 0, 0, 0, 0, 0]
    })
    
    test_data = pd.DataFrame({
        'PassengerId': [7, 8, 9, 10],
        'Pclass': [3, 1, 3, 2],
        'Sex': ['male', 'female', 'female', 'male'],
        'SibSp': [0, 1, 0, 0],
        'Parch': [0, 0, 0, 0]
    })
    print("✅ Sample Titanic dataset created!")
else:
    print("✅ Titanic dataset found! Loading real data...")
    train_data = pd.read_csv(train_path)
    test_data = pd.read_csv(test_path)

# ✅ Step 3: Explore Data (First few rows)
print("\n🔹 Training Data Preview:")
print(train_data.head())

# ✅ Step 4: Prepare Data for Model
y = train_data["Survived"]
features = ["Pclass", "Sex", "SibSp", "Parch"]
X = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])

# ✅ Step 5: Build & Train Random Forest Model
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X, y)

# ✅ Step 6: Make Predictions
predictions = model.predict(X_test)

# ✅ Step 7: Create Submission File
output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
print("\n✅ submission.csv file created successfully!\n")

# ✅ Step 8: Show First Few Rows of Submission File
print("First few rows of your submission file:")
print(output.head())

# ✅ Step 9: Short Project Description
print("\n📘 Project Description:")
print("""
This project predicts passenger survival in the Titanic disaster using a Random Forest Classifier.
The dataset includes features such as passenger class, gender, and family relationships.
After preprocessing categorical data, the model learns survival patterns from the training data
and generates predictions for unseen passengers in the test dataset.
The final output (submission.csv) is ready for Kaggle submission.
""")
```
