```python
# データ読み込み
train_data = pd.read_csv("../input/titanic/train.csv")
# 読み込んだデータの先頭5行を表示（head関数の引数に数字を入れるとその行数分が表示される）
train_data.head()
```

```python
# データ読み込み
test_data = pd.read_csv("../input/titanic/test.csv")
# 読み込んだデータの先頭5行を表示（head関数の引数に数字を入れるとその行数分が表示される）
test_data.head()
```

```python
# 試しに女性の場合の生存率を確認する
# 性別で絞り込みし、列を絞り込む
women = train_data.loc[train_data.Sex == 'female']["Survived"]
# 生存率を計算する（分母：絞り込んだ行数＝女性の総数、分子：生存（1）した人の総数）
rate_women = sum(women)/len(women)
# print data
print("sum : ", sum(women))
print("len : ", len(women))
print(women)
print("% of women who survived : ", rate_women)
```

```python
# 試しに女性の場合の生存率を確認する
# 性別で絞り込みし、列を絞り込む
men = train_data.loc[train_data.Sex == 'male']["Survived"]
# 生存率を計算する（分母：絞り込んだ行数＝男性の総数、分子：生存（1）した人の総数）
rate_men = sum(men)/len(men)
# print data
print("sum : ", sum(men))
print("len : ", len(men))
print(men)
print("% of women who survived : ", rate_men)
```

```python
from sklearn.ensemble import RandomForestClassifier

# 学習データの答え用に列を絞り込む
y = train_data["Survived"]
print(y)

# 学習データの列を絞り込み、カテゴリ変数をダミー変数に変換する
# pandasでカテゴリ変数（カテゴリカルデータ、質的データ）をダミー変数に変換するには、
# pandas.get_dummies()関数を使う。
features = ["Pclass", "Sex", "SibSp", "Parch"]
X = pd.get_dummies(train_data[features])
print(X)

# 学習データと同様に、検査対象データの列を絞り込み、カテゴリ変数をダミー変数に変換する
X_test = pd.get_dummies(test_data[features])
print(X_test)

# モデル作成
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X, y)
# 検査対象データに対して予測する
predictions = model.predict(X_test)

# 予測結果を提出用ファイルに整形する
output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
print(output)
# 提出用ファイルを出力する
output.to_csv('submission.csv', index=False)
print("Your submission was successfully saved!")
```
