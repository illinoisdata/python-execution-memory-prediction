```python
import pandas as pd
import seaborn as sns 
import numpy as np
import matplotlib.pyplot as plt
import missingno as msno
```

```python
train_df=pd.read_csv('/kaggle/input/titanic/train.csv')
test_df=pd.read_csv('/kaggle/input/titanic/test.csv')
```

```python
train_df['Sex'].value_counts()
```

```python
train_df.drop(['PassengerId'], axis=1, inplace=True)
# train_df.drop(['Survived'], axis=1, inplace=True)
train_df.drop(['Ticket'], axis=1, inplace=True)
```

```python
train_df.head()
```

```python
msno.matrix(train_df)
```

```python
train_df.drop(['Cabin'] ,axis=1 , inplace=True)
```

```python
train_df['Age']=train_df['Age'].fillna(train_df.median())
```

```python
train_df['Embarked']=train_df['Embarked'].fillna(train_df.median())
```

```python
train_df['Embarked'].isnull().sum()
```

```python
train_df['Sex']=train_df['Sex'].map({'male':0 , 'female' : 1})
```

```python
train_df=pd.get_dummies(train_df,columns=['Embarked'] , drop_first=True )
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(train_df.corr(), annot=True,     fmt=".2f", cmap='coolwarm', linewidths=0.5,cbar=True )

plt.title('Correlation Heatmap')
plt.show()
```

```python
survivors = train_df[train_df["Survived"] == 1]

# Count survivors by sex (0 = Male, 1 = Female)
sex_counts = survivors["Sex"].value_counts()

# Map labels
labels = sex_counts.index.map({0: "Male", 1: "Female"})

# Plot pie chart
plt.pie(
    sex_counts,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Survivors by Gender")
plt.show()
```
