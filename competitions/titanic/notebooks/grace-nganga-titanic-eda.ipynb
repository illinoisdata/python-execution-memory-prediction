## **Overview**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset (example: Titanic dataset)
df = pd.read_csv('/kaggle/input/titanic/train.csv')

# Preview
df.head()
```

```python
print("Shape of dataset:", df.shape)
df.info()
df.describe()
print("Column names:", df.columns.tolist())
print("Number of unique values:\n", df.nunique())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.isnull().sum().sort_values(ascending=False)
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns
for col in num_cols:
    plt.figure(figsize=(6,3))
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    plt.figure(figsize=(5,3))
    sns.countplot(y=df[col])
    plt.title(f"Count of {col}")
    plt.show()
```

```python
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival Rate by Passenger Class")
plt.show()
```

```python
sns.barplot(x='Sex', y='Survived', data=df)
plt.title("Survival Rate by Gender")
plt.show()
```

```python
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.title("Fare Distribution by Class")
plt.show()
```

```python
sns.scatterplot(x='Age', y='Fare', hue='Survived', style='Pclass', data=df)
plt.title("Age vs Fare vs Survival by Class")
plt.show()

sns.catplot(x='Embarked', y='Survived', hue='Pclass', kind='bar', data=df)
plt.title("Survival Rate by Embarked Location and Class")
plt.show()
```

```python
sns.boxplot(x=df['Fare'])
plt.title("Outliers in Fare")
plt.show()
```

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
upper = Q3 + 1.5*IQR
lower = Q1 - 1.5*IQR

df['Fare_capped'] = np.where(df['Fare'] > upper, upper, 
                             np.where(df['Fare'] < lower, lower, df['Fare']))
```

```python
sns.countplot(x='Survived', data=df)
plt.title("Distribution of Target Variable (Survived)")
plt.show()

df['Survived'].value_counts(normalize=True)
```

```python
sns.boxplot(x='Survived', y='Age', hue='Pclass', data=df)
plt.title("Survival by Age and Class")
plt.show()
```

### **Use Combined Plots to Detect Interaction Effects**

The plot below examines how **Passenger Class (Pclass)** and **Gender (Sex)** interact to influence **Survival (Survived)**.  
By separating the plots by gender (`col="Sex"`) and coloring by survival status (`hue="Survived"`), we can easily observe how survival trends differ across groups.


```python
sns.catplot(x="Pclass", hue="Survived", col="Sex",
            kind="count", data=df, palette="viridis")

plt.subplots_adjust(top=0.8)
plt.suptitle("Survival Count by Passenger Class and Gender", fontsize=14)
plt.show()
```
