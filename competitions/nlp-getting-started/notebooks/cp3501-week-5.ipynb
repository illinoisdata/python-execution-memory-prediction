```python
# Step 1: Imports and Dataset
from fastai.text.all import *
import pandas as pd

# Load train and test datasets
df = pd.read_csv('/kaggle/input/nlp-getting-started/train.csv')
test_df = pd.read_csv('/kaggle/input/nlp-getting-started/test.csv')

# Step 2: DataLoaders
dls = TextDataLoaders.from_df(df, text_col='text', label_col='target', valid_pct=0.2, is_lm=False)

# Step 3: Create and Train the Model
learn = text_classifier_learner(dls, AWD_LSTM, drop_mult=0.5, metrics=accuracy)
learn.fine_tune(4)

# Step 4: Make Predictions
test_dl = dls.test_dl(test_df)
preds = learn.get_preds(dl=test_dl)[0].argmax(dim=1)

# Step 5: Create Submission
submission = pd.DataFrame({'id': test_df['id'], 'target': preds})
submission.to_csv('submission.csv', index=False)
submission.head()
```
