# Data Import and Clean

```python
url = "/kaggle/input/nlp-getting-started/train.csv"
data = pd.read_csv(url, encoding="ISO-8859-1")
data.head()
```

```python
data.shape
```

#### Let's clean the data first

```python
import re
import unicodedata

def clean_text(text):
    # Ensure input is a string
    text = str(text)
    
    # 1. Remove non-standard Unicode characters (commonly causes garbled text)
    text = unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode("utf-8", "ignore")
    
    # 2. Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # 3. Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text, flags=re.MULTILINE)

    # 4. Remove non-alphabetic characters (optional)
    text = re.sub(r"[^a-zA-Z\s]", "", text)

    # 5. Convert text to lowercase
    text = text.lower()
    
    # 6. Remove extra whitespaces
    text = re.sub(r"\s+", " ", text).strip()

    return text

data["clean_text"] = data["text"].apply(clean_text)
data["clean_text"]
```

#### Create a TF-IDF instance.

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    ngram_range=(1, 2)
)

vec_model = vectorizer.fit_transform(data["clean_text"])
vec_model
```

#### Examine the vocabulary 

```python
sorted(vectorizer.vocabulary_.items(), key=lambda item:item[1])[0:10]
```

#### "aa" seems weird and exist in many tweets, let's try to examine it.

```python
for i in range(100):
    if "aa" in data.iloc[i,-1]:
        print(data.iloc[i,-1])
        print(data.iloc[i,-3])
```

#### It appears "aa" is some kind of garbled character that's causing errors after processing. We should remove it.

```python
data["clean_text"] = data["clean_text"].apply(lambda text : re.sub(r'\b(a)\1{1,}\b', '', text))
data["clean_text"]
```

#### Examine the model again

```python
vectorizer.fit_transform(data["clean_text"])
sorted(vectorizer.vocabulary_.items(), key=lambda item:item[1])[0:10]
```

```python
idf_scores = dict(zip(vectorizer.get_feature_names_out(), vectorizer.idf_))
#see the idf score for each word (output shows only some of them)
idf_scores
```

#### Import pre-trained FastText model

```python
import fasttext

fasttext_model = fasttext.load_model("/kaggle/input/cc.en.300.bin/other/default/1/cc.en.300.bin")
```

#### Calculating the weighted vecs

```python
def tfidf_weighted_embedding(text):
    words = text.split()
    vecs = []
    weights = []
    
    for word in words:
        if word in idf_scores:
            vec = fasttext_model.get_word_vector(word)
            weight = idf_scores[word]
            vecs.append(vec * weight)
            weights.append(weight)
        else:
            vec = fasttext_model.get_word_vector(word)
            weight = 1
            vecs.append(vec * weight)
            weights.append(weight)

    if vecs:
        # divide the sum of the weights for scaling
        return np.sum(vecs, axis=0) / np.sum(weights)
    else:
        return np.zeros(fasttext_model.get_dimension())


data["embedding_vector"] = data["clean_text"].apply(tfidf_weighted_embedding)


embedding_df = pd.DataFrame(data["embedding_vector"].tolist(), columns=[f"weight_vector{i}" for i in range(1, 301)])
embedding_df.head()
```

```python
embedding_df.shape
```

# Model Training

#### Train test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(embedding_df, data["target"], test_size=0.2, random_state=69)  
```

#### Using the function in the last notebook for NLP.

```python
from sklearn.model_selection import KFold, cross_val_score, StratifiedKFold, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
import time

def model_experiment(grid_para, classifier_ins, df):
    """
    Performs experiments for regression models, including feature selection, preprocessing, grid search, and performance evaluation.
    
    Args:
        grid_para (dict): A dictionary containing parameters for grid search.
        regressor_ins (sklearn.base.RegressorMixin): An instance of a regression model (e.g., LinearRegression(), XGBRegressor()).
        df_results (pd.DataFrame): The DataFrame used to store results.
    
    Returns:
        pd.DataFrame: The updated results DataFrame.
    """
    #use the parameters 
    grid_para = grid_para
    #create k-fold 
    kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=43)
    
    #create grid search
    grid_search = GridSearchCV(estimator=classifier_ins,
                           param_grid=grid_para,
                           cv=kfold,
                           scoring="accuracy",
                           n_jobs=-1,
                           verbose=1)
    
    #find the best model and count time
    start_time = time.time()

    
    # start model
    grid_search.fit(X_train, y_train)

    # grid_search.fit(X_train, y_train)
    best_model=grid_search.best_estimator_
    
    #print the score
    print("\nGridSearchCV execution completed.")
    print(f"Best hyperparameters: {grid_search.best_params_}")
    print("--------------------------------------------------------------------------")
    print(f"Train Accuracy: {best_model.score(X_train, y_train):}")
    print(f"Average accuracy of the best model during cross-validation: {grid_search.best_score_:}")
    print(f"Final Test Accuracy = {best_model.score(X_test, y_test):}")

    end_time = time.time()
    training_time = round(end_time - start_time, 2)
    
    #put in the dataframe
    new_row={        
        "Best hyperparameters":grid_search.best_params_, 
        "Train Accuracy": round(best_model.score(X_train, y_train), 4),
        "Cross-validation Accuracy" : round(grid_search.best_score_ ,4),
        "Final Test Accuracy": round(best_model.score(X_test, y_test), 4),
        "Best Model": best_model,
        "Training Time":training_time
        }
    df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)

    #draw the confusion matrix
    
    y_predict=best_model.predict(X_test)
    
    cm = confusion_matrix(y_test, y_predict)
    ax= plt.subplot()
    sns.heatmap(cm, annot=True, ax = ax, fmt="d"); 
    ax.set_xlabel("Predicted labels")
    ax.set_ylabel("True labels")
    ax.set_title("Confusion Matrix"); 
    ax.xaxis.set_ticklabels(["Not Disaster", "Disaster"]); ax.yaxis.set_ticklabels(["Not Disaster", "Disaster"]) 
    plt.show() 

    return df
```

#### Initialize the classfier

```python
import xgboost as xgb

xgb_results = pd.DataFrame(columns=[        
    "Best hyperparameters",     
    "Train Accuracy",
    "Cross-validation Accuracy",         
    "Final Test Accuracy", 
    "Best Model",
    "Training Time"
])
xgb_class = xgb.XGBClassifier(random_state=42, use_label_encoder=False, eval_metric="logloss")
```

```python
grid_para= {
    "n_estimators": [100, 200], 
    "learning_rate": [0.05, 0.1], 
    "max_depth": [3, 5], 
    "subsample": [0.6, 0.7], 
    "colsample_bytree": [0.6, 0.7], 
    "reg_alpha": [0, 0.5], 
    "reg_lambda": [1, 3] 
}

xgb_results = model_experiment(grid_para, xgb_class, xgb_results)
xgb_results
```

```python
grid_para= {
    "n_estimators": [50, 80], 
    "learning_rate": [0.05, 0.1], 
    "max_depth": [1, 3, 5], 
    "subsample": [0.6, 0.7], 
    "colsample_bytree": [0.6, 0.7], 
    "reg_alpha": [0, 0.5], 
    "reg_lambda": [1, 3] 
}

xgb_results = model_experiment(grid_para, xgb_class, xgb_results)
xgb_results
```

#### Utilize the early-stopping to reduce overfitting.

```python
best_params = xgb_results.iloc[-1, 0]

# Set the best paramater in the best model.
final_model = xgb.XGBClassifier(
    **best_params,
    random_state=42,
    use_label_encoder=False,
    eval_metric="logloss",
    early_stopping_rounds=10
    )

X_train_sub, X_valid, y_train_sub, y_valid = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
    )

#Fit the model
final_model.fit(
    X_train_sub, y_train_sub,
    eval_set=[(X_valid, y_valid)],
    verbose=False
    )

# Evaluation
train_acc = final_model.score(X_train, y_train)
test_acc = final_model.score(X_test, y_test)

print(f"Train Accuracy: {train_acc:.4f}")
print(f"Final Test Accuracy = {test_acc:.4f}")
```
