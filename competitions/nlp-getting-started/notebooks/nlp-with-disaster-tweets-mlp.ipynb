# For this jupyter notebook approach
Using Multi-Layer Perceptrons (MLPs) from scikit-learn for Natural Language Processing (NLP) involves several key steps to prepare text data for the neural network.
1. Text Preprocessing:
Tokenization: Divide text into individual words or subword units (tokens).
Lowercasing: Convert all text to lowercase to standardize words.
Stop Word Removal: Eliminate common words (e.g., "the," "a," "is") that often carry little semantic meaning.
Stemming or Lemmatization: Reduce words to their root form (e.g., "running," "runs," "ran" to "run").
2. Feature Extraction:
Vectorization: Convert processed text into numerical representations that the MLP can understand. Common methods include:
CountVectorizer: Creates a matrix where each row represents a document and each column represents a word, with values indicating the frequency of that word in the document.
TfidfVectorizer: Similar to CountVectorizer but also considers the inverse document frequency, giving less weight to common words and more weight to rare, informative words.
3. MLP Model Training with MLPClassifier or MLPRegressor:
import.

```Python
from sklearn.neural_network import MLPClassifier
```
or

```Python
from sklearn.neural_network import MLPRegressor
```
Initialization: Create an instance of the MLP model, specifying parameters such as:
hidden_layer_sizes: A tuple defining the number of neurons in each hidden layer (e.g., (100, 50) for two hidden layers with 100 and 50 neurons, respectively).
activation: The activation function (e.g., 'relu', 'tanh', 'logistic').
solver: The optimization algorithm (e.g., 'adam', 'sgd', 'lbfgs').
max_iter: The maximum number of training iterations.
random_state: For reproducibility.
Training: Fit the model to your vectorized training data and corresponding labels (for classification) or target values (for regression).
Python

    model.fit(X_train_vectorized, y_train)
4. Prediction and Evaluation:
Prediction: Use the trained model to make predictions on new, unseen vectorized text data.
Python
    ```python
    predictions = model.predict(X_test_vectorized)
    ```
Evaluation: Assess the model's performance using appropriate metrics (e.g., accuracy, precision, recall, F1-score for classification; mean squared error for regression).

# Example Workflow (Classification):
```Python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Sample text data and labels
texts = ["This is a positive review.", "Terrible service, very bad.", "Good product, happy with it.", "Worst experience ever."]
labels = [1, 0, 1, 0] # 1 for positive, 0 for negative

# 1. & 2. Text Preprocessing and Feature Extraction
vectorizer = TfidfVectorizer(stop_words='english')
X_vectorized = vectorizer.fit_transform(texts)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_vectorized, labels, test_size=0.5, random_state=42)

# 3. MLP Model Training
mlp = MLPClassifier(hidden_layer_sizes=(100,), activation='relu', solver='adam', max_iter=200, random_state=42)
mlp.fit(X_train, y_train)

# 4. Prediction and Evaluation
y_pred = mlp.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")
```

#  Fine Tuning TFIDF performance
### 1. Enhance Data Quality and Quantity

Clean and preprocess text more effectively: Your TF-IDF likely uses raw text, but adding steps like lowercase conversion, punctuation removal, stop word elimination, stemming, or lemmatization can reduce noise. Use libraries like NLTK or spaCy for this.

Example: Before vectorizing, apply a preprocessing function to your text data.


Handle class imbalance: If your labels are imbalanced (check with value_counts() on y_train), F1 can suffer on minority classes. Use techniques like oversampling (SMOTE from imbalanced-learn), undersampling, or class weights in MLPClassifier (set class_weight='balanced').
Augment data: If your dataset is small, generate synthetic examples using synonyms (via WordNet) or back-translation. Or collect more labeled data if feasible.
Increase dataset size/split properly: Ensure your train-validation split is stratified (use train_test_split with stratify=y). Try k-fold cross-validation (e.g., via StratifiedKFold) to get a more reliable F1 estimate.

### 2. Optimize Feature Extraction (TF-IDF)
TF-IDF is solid for text, but tuning it can boost performance by capturing better n-grams or filtering irrelevant terms.

Tune TF-IDF parameters: Use GridSearchCV to experiment with:



ParameterSuggestions to TryWhy It Helpsngram_range(1,1) → (1,2) or (1,3)Captures phrases like "machine learning" instead of single words.max_featuresNone → 5000, 10000, or 20000Limits vocabulary to top terms, reducing noise and dimensionality.min_df / max_dfmin_df=2-5, max_df=0.8-0.95Ignores rare or overly common words that don't discriminate classes.

Fit the vectorizer only on x_train (to avoid data leakage), then transform x_val.


Combine with other features: Add bigrams/trigrams or even simple features like text length if relevant to your task.
Alternative vectorizers: Try CountVectorizer instead of TF-IDF for comparison, or combine both via feature union in a pipeline.

### 3. Tune MLPClassifier Hyperparameters
MLP's defaults might not be optimal; use hyperparameter search to find better configs.

Key parameters to tune (via GridSearchCV or RandomizedSearchCV on a Pipeline of TfidfVectorizer + MLPClassifier):



ParameterSuggestions to TryWhy It Helpshidden_layer_sizes(100,) → (50,50), (100,50), (200,)Deeper/wider layers can capture complex patterns in text vectors. Start small to avoid overfitting.activation'relu' → 'logistic', 'tanh'Different non-linearities might suit your data better.solver'adam' → 'sgd', 'lbfgs''adam' is good for large data; 'lbfgs' for small.alpha0.0001 → 0.001, 0.01, 0.1Increases regularization to prevent overfitting.learning_rate_init0.001 → 0.01, 0.0001 (if solver='adam' or 'sgd')Controls training speed; too high causes instability.max_iter200 → 500 or moreEnsures convergence if early stopping isn't used.early_stoppingFalse → True (with validation_fraction=0.1)Stops training when validation score plateaus, reducing overfitting.

# Split Train/Test dataset

````python
X_train, X_test, y_train, y_test =  train_test_split(combinedTfid[: trainDataSize], trainDF["target"], 
                                                     test_size=0.2, random_state=42 ,stratify=trainDF["target"])
````

### MLP inital

````python
mlp = MLPClassifier(hidden_layer_sizes=(2048, 1024, 512), #(256,128,64,32),
                    activation= "logistic", #"relu", 
                    solver="adam",
                    
                    verbose=1,
                    
                    )
mlp
````

````python
%%time
start= time.time()
mlp.fit(X_train, y_train)
print(f"Train Time: {time.time() - start} seconds")
````

````python
# 4. Plot the loss curve
loss_values= mlp.loss_curve_
plt.figure(figsize=(8, 6))
plt.plot(loss_values)
plt.title('MLP Training Loss Curve')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)
plt.show()
````
