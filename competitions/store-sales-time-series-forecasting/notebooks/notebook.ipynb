```python
import pandas as pd

# Load datasets
train = pd.read_csv("/kaggle/input/store-sales-time-series-forecasting/train.csv")
test = pd.read_csv("/kaggle/input/store-sales-time-series-forecasting/test.csv")

# Ensure no negative sales values (clip to 0)
train['sales'] = train['sales'].clip(lower=0)

# Compute average sales per (store_nbr, family)
avg_sales = train.groupby(['store_nbr', 'family'])['sales'].mean().reset_index()

# Merge average sales with test set on store and family
submission = pd.merge(test, avg_sales, on=['store_nbr', 'family'], how='left')

# If any values are missing (store-family combo never seen), fill with global average
global_avg = train['sales'].mean()
submission['sales'] = submission['sales'].fillna(global_avg)


# Prepare final submission file
submission = submission[['id', 'sales']]
submission.to_csv("submission.csv", index=False)



```
