#### columns that doesn't help in identifying the cancer type

```python
column = ['smoothness_mean','symmetry_worst','symmetry_mean','fractal_dimension_se','smoothness_se'
          ,'fractal_dimension_worst','fractal_dimension_mean','texture_se','compactness_se','smoothness_worst']
```

```python
train.drop(columns = column, axis = 1 , inplace = True)
test.drop(columns = column , axis = 1, inplace = True)
```

```python
test.columns
```

```python
ids = test['id']
```

```python
train.drop('id', axis = 1 , inplace = True)
test.drop('id', axis = 1 , inplace = True)
```

```python
train['diagnosis'] = train['diagnosis'].replace({'M':1,'B':0})
```

```python
X = train.drop('diagnosis',axis = 1)
y = train['diagnosis']
```

```python
train.info()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.1,random_state=23,stratify=y)
```

```python
def scalling(df) :
    scalerstandard = StandardScaler()
    x1 = scalerstandard.fit_transform(df)
    x2 = pd.DataFrame(x1,columns = df.columns)
    return x2
```

```python
X_train = scalling(X_train)
```

```python
X_test = scalling(X_test)
```

```python
KNN = KNeighborsClassifier(n_neighbors=5,metric='cosine')
KNN.fit(X_train,y_train)
KNN.score(X_test,y_test)
```

## training the model over all the data

### KNN

```python
KNN.fit(X,y)
```

```python
y_pred = KNN.predict(test)
```

```python
submission_1 = pd.DataFrame({
    'id' : ids,
    'diagnosis' : y_pred
})
```

```python
submission_1['diagnosis'] = submission_1['diagnosis'].replace({1:'M',0:'B'})
```

```python
submission_1.to_csv('/kaggle/working/submission_1',index=False)
```
