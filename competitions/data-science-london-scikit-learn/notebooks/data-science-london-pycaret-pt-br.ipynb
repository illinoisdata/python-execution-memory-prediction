```python
from pycaret.classification import *
```

# <font color='blue' style='font-size: 30px;'>PYCARET</font>


```python
# Setup
setup_de_classificacao = setup(data=train_data_new, target='result')
```

```python
# Compare Models
melhores_modelos = compare_models(n_select=5)
```

```python
melhores_modelos[0]
```

```python
type(melhores_modelos[0])
```

```python
# K Neighbors Classifier - Tunagem de hiperparametros do KNN
knn_tunado = tune_model(melhores_modelos[0])
```

```python
# Finalizar o modelo
modelo_final = finalize_model(knn_tunado)
```

```python
modelo_final
```

```python
plot_model(modelo_final)
```

# <font color='blue' style='font-size: 30px;'>Predições</font>


```python
predicoes = predict_model(modelo_final, test_data)
predicoes
```

```python
df_final = test_data.reset_index()
df_final
```

# <font color='blue' style='font-size: 30px;'>Submit to Kaggle</font>


```python
submission = pd.DataFrame( {'id': np.arange(1,df_final.shape[0]+1), 'Solution' : predicoes["prediction_label"]})

submission.to_csv('submission.csv', index=False)
```
