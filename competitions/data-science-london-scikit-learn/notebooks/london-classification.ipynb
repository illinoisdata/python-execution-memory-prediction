# Read Data

```python
train_data = pd.read_csv('/kaggle/input/data-science-london-scikit-learn/train.csv',header=None)
train_data.head()
```

```python
trainlabel_data = pd.read_csv('/kaggle/input/data-science-london-scikit-learn/trainLabels.csv',header=None)
trainlabel_data.head()
```

# Classification before Pre-Processing

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(train_data, trainlabel_data, test_size=0.20,
                                                    random_state=42)
print(X_train.shape,y_train.shape)
print(X_test.shape,y_test.shape)
```

# Models ML

```python
from sklearn.metrics import accuracy_score
```

```python
# DecisionTreeClassifier
from sklearn.tree import DecisionTreeClassifier
clf_tree = DecisionTreeClassifier(criterion='entropy',max_depth=10000,max_leaf_nodes=1000)
clf_tree.fit(X_train,y_train)
y_pred_Dtree = clf_tree.predict(X_test)
accuracy_DTC = accuracy_score(y_test,y_pred_Dtree)
print('DecisionTreeClassifier',accuracy_DTC)
```

```python
# RandomForestClassifier
from sklearn.ensemble import RandomForestClassifier
clf_rfc = RandomForestClassifier(criterion='entropy')
clf_rfc.fit(X_train,y_train)
y_pred_rfc = clf_rfc.predict(X_test)
accuracy_RFC = accuracy_score(y_test,y_pred_rfc)
print('RandomForestClassifier',accuracy_RFC)
```

```python
# Gaussian Naive Bayes  
from sklearn.naive_bayes import GaussianNB
clf_GNB = GaussianNB()
clf_GNB.fit(X_train,y_train)
y_pred_GNB = clf_GNB.predict(X_test)
accuracy_GNB = accuracy_score(y_test,y_pred_GNB)
print('Gaussian Naive Bayes ' , accuracy_GNB)
```

```python
# Linear Support Vectoer Classifier(SVC)
from sklearn.svm import LinearSVC
clf_L_svc = LinearSVC()
clf_L_svc.fit(X_train,y_train)
y_pred_L_SVC = clf_L_svc.predict(X_test)
accuracy_L_SVC = accuracy_score(y_test,y_pred_L_SVC)
print('Linear Support Vectoer Classifier' , accuracy_L_SVC)
```

```python
# Support Vectoer Classifier(SVC)
from sklearn.svm import SVC
clf_svc = SVC(C=3)
#clf_svc = SVC(C=3)
clf_svc.fit(X_train,y_train)
y_pred_SVC = clf_svc.predict(X_test)
accuracy_SVC = accuracy_score(y_test,y_pred_SVC)
print(' Support Vectoer Classifier' , accuracy_SVC)
```

```python
# LogisticRegression
from sklearn.linear_model import LogisticRegression
clf_LR = LogisticRegression()
clf_LR.fit(X_train,y_train)
y_pred_LR = clf_LR.predict(X_test)
accuracy_LR = accuracy_score(y_test,y_pred_LR)
print('LogisticRegression Classifier ' , accuracy_LR)
```

```python
#  k-Nearest Neighbors(KNN)
from sklearn.neighbors import KNeighborsClassifier
clf_KNN = KNeighborsClassifier(n_neighbors=5)
clf_KNN.fit(X_train,y_train)
y_pred_knn = clf_KNN.predict(X_test)
accuracy_KNN = accuracy_score(y_test,y_pred_knn)
print('k-Nearest Neighbors(KNN)' , accuracy_KNN)
```

```python
models = pd.DataFrame({
    'Models' : ['KNeighborsClassifier','LogisticRegression','Support Vectoer Classifier',
                'Linear Support Vectoer Classifier','GaussianNB','RandomForestClassifier',
                'DecisionTreeClassifier'],
    'Accuracy' : [accuracy_KNN , accuracy_LR ,accuracy_SVC,
                 accuracy_L_SVC,accuracy_GNB , accuracy_RFC , accuracy_DTC ]
})
models =models.sort_values(by='Accuracy',ascending=False)
models
```

```python
sns.barplot(y='Models',x='Accuracy',data= models)
```

```python
# RandomForestClassifier
val_RF = RandomForestClassifier()  # model
score_RF = cross_val_score(val_RF , train_data, trainlabel_data,cv=5)
print('RandomForestClassifier using Cross Validation : ',score_RF.mean())
```

# GridSearchCV , RandomizedSearchCV

```python
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV
```

# Random Forest Classifier USING GRID SEARCH

```python
# Random Forest Classifier

clf_RF = RandomForestClassifier()

best_params = { 'n_estimators' : np.arange(100,400,50) ,
                'criterion' :['entropy','gini'],
                'max_depth' : np.arange(5,30,5),
                'max_samples':[0.7,0.8,0.9,1]}

# USING GRID SEARCH

gridsearch_RF = GridSearchCV(estimator=clf_RF,param_grid=best_params,cv=5,n_jobs=-1)
gridsearch_RF.fit(X_train,y_train)
y_pred_RF_grid = gridsearch_RF.predict(X_test)
accurcy_GS_RF = accuracy_score(y_test,y_pred_RF_grid)
print('Random Forest Classifier using GridSearchCV',accurcy_GS_RF)
```

```python
gridsearch_RF.best_params_
```

```python
gridsearch_RF.best_estimator_
```

```python
best_Random_forest_grid = gridsearch_RF.best_estimator_
best_Random_forest_grid.fit(X_train,y_train)
y_pred_best_Rf_grid = best_Random_forest_grid.predict(X_test)
accuracy_grid_rf = accuracy_score(y_test,y_pred_best_Rf_grid)
print('Random Forest Classifier using GridSearch ' , accuracy_grid_rf)
```

# k-Nearest Neighbors(KNN)Classifier USING RandomizedSearchCV

```python
#  k-Nearest Neighbors(KNN)

clf_KNN_GS = KNeighborsClassifier()

best_params = { 'n_neighbors' : np.arange(3,13,2) ,
                'weights' : ['uniform', 'distance'],
                'metric' : ['minkowski','Euclidean '],
                'p' : [2,3]}

# USING RandomizedSearchCV

randomsearch_knn = RandomizedSearchCV(estimator=clf_KNN_GS,param_distributions=best_params,cv=5,n_jobs=-1)
randomsearch_knn.fit(X_train,y_train)
y_pred_KNN_RS = randomsearch_knn.predict(X_test)
accurcy_RS_knn = accuracy_score(y_test,y_pred_KNN_RS)
print('k-Nearest Neighbors(KNN) Classifier using RandomizedSearchCV',accurcy_RS_knn)
```

```python
randomsearch_knn.best_params_
```

```python
# Model
randomsearch_knn.best_estimator_
```

```python
best_knn_RS = randomsearch_knn.best_estimator_
best_knn_RS.fit(X_train,y_train)
y_pred_best_knn_Rs = best_knn_RS.predict(X_test)
accuracy_Rs_knn = accuracy_score(y_test,y_pred_best_knn_Rs)
print(' k-Nearest Neighbors(KNN) Classifier' , accuracy_Rs_knn)
```

# Support Vectoer Classifier(SVC) USING GRID SEARCH

```python
# Support Vectoer Classifier(SVC)

#from sklearn.svm import SVC
clf_SVC_GS = SVC()

best_params = { 'C' : [1.0,10,50,100,150,200,250] ,
                'kernel' : ['poly','sigmoid'],
                'degree' : [2,3,4,5,6,7],
                'gamma' : ['auto','scale']}

# USING RandomizedSearchCV

randomsearch_svc = RandomizedSearchCV(estimator=clf_SVC_GS,param_distributions=best_params,cv=5,n_jobs=-1)
randomsearch_svc.fit(X_train,y_train)
y_pred_svc_RS = randomsearch_svc.predict(X_test)
accurcy_RS_svc = accuracy_score(y_test,y_pred_svc_RS)
print('Support Vectoer Classifier(SVC) using RandomizedSearchCV',accurcy_RS_svc)
```

```python
# Support Vectoer Classifier(SVC)

# from sklearn.svm import SVC
clf_SVC_GS = SVC()

best_params = { 'C' : [0.001,0.1,1,2,3,4,5,10,20,30,40,50,100,150,200,250] ,
                'kernel' : ['poly','sigmoid'],
                'degree' : [2,3,4,5,6,7],
                'gamma' : ['auto','scale']}

# USING RandomizedSearchCV

gridsearch_svc = GridSearchCV(estimator=clf_SVC_GS,param_grid=best_params,cv=5,n_jobs=-1)
gridsearch_svc.fit(X_train,y_train)
y_pred_svc_GS = gridsearch_svc.predict(X_test)
accurcy_GS_svc = accuracy_score(y_test,y_pred_svc_GS)
print('Support Vectoer Classifier(SVC) using GridizedSearchCV',accurcy_GS_svc)
```

```python
best_svc_GS = gridsearch_svc.best_estimator_
best_svc_GS.fit(X_train,y_train)
y_pred_best_svc_GS = best_svc_GS.predict(X_test)
accuracy_Gs_svc = accuracy_score(y_test,y_pred_best_svc_GS)
print(' Support Vectoer Classifier(SVC)' , accuracy_Gs_svc)
```
