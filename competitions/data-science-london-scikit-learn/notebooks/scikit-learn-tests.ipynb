```python
train_df = pd.read_csv('/kaggle/input/data-science-london-scikit-learn/train.csv', header = None)
test_df = pd.read_csv('/kaggle/input/data-science-london-scikit-learn/test.csv', header = None)
trainLabels = pd.read_csv('/kaggle/input/data-science-london-scikit-learn/trainLabels.csv', header = None)
combine = [train_df, test_df]
train_df.shape, test_df.shape
```

```python
decision_tree = DecisionTreeClassifier()
decision_tree.fit(train_df, trainLabels.values.ravel())
decision_tree_out= decision_tree.predict(test_df)
acc_decision_tree = round(decision_tree.score(train_df, trainLabels) * 100, 2)
acc_decision_tree
```

```python
logistic_regression = LogisticRegression()
logistic_regression.fit(train_df, trainLabels.values.ravel())
logistic_regression_out= logistic_regression.predict(test_df)
acc_logistic_regression = round(logistic_regression.score(train_df, trainLabels) * 100, 2)
acc_logistic_regression
```

```python
gradient_boosting= GradientBoostingClassifier()
gradient_boosting.fit(train_df, trainLabels.values.ravel())
gradient_boosting_out= gradient_boosting.predict(test_df)
acc_gradient_boosting = round(gradient_boosting.score(train_df, trainLabels) * 100, 2)
acc_gradient_boosting
```

```python
random_forest = RandomForestClassifier()
random_forest.fit(train_df, trainLabels.values.ravel())
random_forest_out = random_forest.predict(test_df)
acc_random_forest = round(random_forest.score(train_df, trainLabels) * 100, 2)
acc_random_forest
```

```python
knn = KNeighborsClassifier()
knn.fit(train_df, trainLabels.values.ravel())
knn_out = random_forest.predict(test_df)
acc_knn = round(knn.score(train_df, trainLabels) * 100, 2)
acc_knn
```

```python
models = pd.DataFrame({
    'Model': ['Decision Tree', 'Logistic Regression', 'Gradient Boosting Classifier',
              'Random Forest Classifier', 'KNeighbors Classifier'],
    'Score': [acc_decision_tree, acc_logistic_regression, acc_gradient_boosting, 
              acc_random_forest, acc_knn]})
models.sort_values(by='Score', ascending=False)
```
