```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv('/kaggle/input/data-science-london-scikit-learn/train.csv',header=None)
df1=pd.read_csv('/kaggle/input/data-science-london-scikit-learn/test.csv',header=None)
label=pd.read_csv('/kaggle/input/data-science-london-scikit-learn/trainLabels.csv',header=None)
df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df,label,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import LogisticRegression
model1=LogisticRegression()
model1.fit(x_train,y_train)
```

```python
y_pred=model1.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print('Accuracy:',accuracy_score(y_test,y_pred))
```

```python
y_train=y_train.values.ravel()
from sklearn.model_selection import cross_val_score
score=cross_val_score(model1,x_train,y_train,cv=5,scoring='accuracy').mean()
print(score)
```

```python
params={
    'C':[0.001,0.01,0.1,1,10,100]
}
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(model1,params,scoring='accuracy')
grid.fit(x_train,y_train)
```

```python
grid.best_score_,grid.best_estimator_,grid.best_params_
```

Since we are getting the accuracy of 82% only we will try different classification algorithms

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
knn=KNeighborsClassifier()
Rf=RandomForestClassifier()
```

```python
# Grid Search for best parameters for KNN
knn_grid=GridSearchCV(knn,param_grid={},cv=5,scoring='accuracy').fit(x_train,y_train)
knn_grid.best_score_,knn_grid.best_estimator_
```

```python
# Grid Search for best parameters for Random forest
Rf_grid=GridSearchCV(Rf,param_grid={},cv=5,scoring='accuracy').fit(x_train,y_train)
Rf_grid.best_score_,Rf_grid.best_estimator_
```

```python
knn.fit(x_train,y_train)
```

```python
y_pred=knn.predict(x_test)
print('Accuracy:',accuracy_score(y_test,y_pred))
sc=cross_val_score(knn,x_train,y_train,cv=5,scoring='accuracy').mean()
print('Cross validation score:',sc)
```

```python
Rf.fit(x_train,y_train)
y_pred=Rf.predict(x_test)
print('Accuracy:',accuracy_score(y_test,y_pred))
sc=cross_val_score(Rf,x_train,y_train,cv=5,scoring='accuracy').mean()
print('Cross validation score:',sc)
```

```python
from xgboost import XGBClassifier
Xg=XGBClassifier()
Xg.fit(x_train,y_train)
y_pred=Xg.predict(x_test)
print('Accuracy:',accuracy_score(y_test,y_pred))
sc=cross_val_score(Xg,x_train,y_train,cv=5,scoring='accuracy').mean()
print('Cross validation score:',sc)
```

```python
from catboost import CatBoostClassifier
cb=CatBoostClassifier()
cb.fit(x_train,y_train)
y_pred=cb.predict(x_test)
print('Accuracy:',accuracy_score(y_test,y_pred))
sc=cross_val_score(cb,x_train,y_train,cv=5,scoring='accuracy').mean()
print('Cross validation score:',sc)
```

KNN had better accuracy among all
