```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import train_test_split
%matplotlib inline
```

```python
df = pd.read_csv("/kaggle/input/data-science-london-scikit-learn/train.csv",header=None)
df
```

```python
l = pd.read_csv("/kaggle/input/data-science-london-scikit-learn/trainLabels.csv",header=None)
l.head()
```

```python
df.isnull().sum()
```

```python
x,y = df,l
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.1,random_state=42)
```

```python
from sklearn.svm import SVC

m = SVC()
```

```python
m.fit(x_train,y_train.values.ravel())
m.score(x_test,y_test)
```

```python
m.score(x_train,y_train)
```

```python
from sklearn.metrics import classification_report
pred = m.predict(x_test)
print(m.__class__.__name__)
print(classification_report(y_test,pred))
```

```python
dft = pd.read_csv("/kaggle/input/data-science-london-scikit-learn/test.csv",header=None)
dft.head()
```

```python
df1 = pd.DataFrame({'id':np.arange(1,9001,1), 'Solution':np.arange(1,9001,1)})
df1.head()
pre = m.predict(dft)
df1['Solution']=pre
df1.to_csv("Ans.csv",index=False)
```
