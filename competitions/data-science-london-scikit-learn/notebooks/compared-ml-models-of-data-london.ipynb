```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_data = pd.read_csv('../input/data-science-london-scikit-learn/train.csv',header = None)
train_labels = pd.read_csv('../input/data-science-london-scikit-learn/trainLabels.csv',header = None)
test_data =  pd.read_csv('../input/data-science-london-scikit-learn/test.csv',header = None)
```

# ML_Models 

```python
from sklearn.model_selection import cross_val_score, train_test_split

X, y = train_data, np.ravel(train_labels)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

# KNN

```python
# KNN with cross-validation
from sklearn.neighbors import KNeighborsClassifier

# Model complexity
neig = np.arange(1, 25)
kfold = 10
train_accuracy = []
val_accuracy = []
bestKnn = None
bestAcc = 0.0
# Loop over different values of k
for i, k in enumerate(neig):
    # k from 1 to 25(exclude)
    knn = KNeighborsClassifier(n_neighbors=k)
    # Fit with knn
    knn.fit(X_train,y_train)
    #train accuracy
    train_accuracy.append(knn.score(X_train, y_train))
    # test accuracy
    val_accuracy.append(np.mean(cross_val_score(knn, X, y, cv=kfold)))
    if np.mean(cross_val_score(knn, X, y, cv=kfold)) > bestAcc:
        bestAcc = np.mean(cross_val_score(knn, X, y, cv=10))
        bestKnn = knn
        
        
# Plot
plt.figure(figsize=[13,8])
plt.plot(neig, val_accuracy, label = 'Validation Accuracy')
plt.plot(neig, train_accuracy, label = 'Training Accuracy')
plt.legend()
plt.title('k value VS Accuracy')
plt.xlabel('Number of Neighbors')
plt.ylabel('Accuracy')
plt.xticks(neig)
plt.show()

print('Best Accuracy without feature scaling:', bestAcc)
print(bestKnn)
```

# BernoulliNB

```python
from sklearn.naive_bayes import BernoulliNB
# Model complexity
alpha = np.arange(0.1, 50)
train_accuracy = []
val_accuracy = []
bestBernoulliNB = None
bestAcc = 0.0
# Loop over different values of k
for i, k in enumerate(alpha):
    # k from 0.1 to 1(exclude)
    BernoulliNBModel = BernoulliNB(alpha=k,binarize=1)
    # Fit with BernoulliNBModel
    BernoulliNBModel.fit(X_train,y_train)
    #train accuracy
    train_accuracy.append(BernoulliNBModel.score(X_train, y_train))
    # test accuracy
    val_accuracy.append(np.mean(cross_val_score(BernoulliNBModel, X, y)))
    if np.mean(cross_val_score(BernoulliNBModel, X, y)) > bestAcc:
        bestAcc = np.mean(cross_val_score(BernoulliNBModel, X, y))
        bestBernoulliNB = BernoulliNBModel
# Plot
plt.figure(figsize=[13,8])
plt.plot(alpha, val_accuracy, label = 'Validation Accuracy')
plt.plot(alpha, train_accuracy, label = 'Training Accuracy')
plt.legend()
plt.title('alpha value VS Accuracy')
plt.xlabel('Number of alpha')
plt.ylabel('Accuracy')
plt.xticks(alpha)
plt.show()

print('Best Accuracy without feature scaling:', bestAcc)
print(bestBernoulliNB)
```

# Gradient Boosting Regressor

```python
#GradientBoostingRegressor
from sklearn.ensemble import GradientBoostingRegressor
max_depth=np.arange(1,11)
train_accuracy = []
val_accuracy = []
bestGradientBoostingRegressor = None
bestAcc = 0.0
# Loop over different values of m
for l , m in enumerate(max_depth):
        # m from 1 to 11 (exclude)
        GradientBoostingRegressorModel = GradientBoostingRegressor(n_estimators=100,max_depth=m, random_state=33)
        # Fit with RandomForestRegressor
        GradientBoostingRegressorModel.fit(X_train,y_train)
        #train accuracy
        train_accuracy.append(GradientBoostingRegressorModel.score(X_train, y_train))
        # test accuracy
        val_accuracy.append(np.mean(cross_val_score(GradientBoostingRegressorModel, X, y)))
        if np.mean(cross_val_score(GradientBoostingRegressorModel, X, y)) > bestAcc:
            bestAcc = np.mean(cross_val_score(GradientBoostingRegressorModel, X, y))
            bestGradientBoostingRegressor = GradientBoostingRegressorModel
    
# Plot
plt.figure(figsize=[13,8])
plt.plot(max_depth, val_accuracy, label = 'Validation Accuracy')
plt.plot(max_depth, train_accuracy, label = 'Training Accuracy')
plt.legend()
plt.title('max_depth VS Accuracy')
plt.xlabel('Number of max_depth')
plt.ylabel('Accuracy')
plt.xticks(max_depth)

plt.show()

print('Best Accuracy without feature scaling:', bestAcc)
print(bestGradientBoostingRegressor)
```

```python
from sklearn.preprocessing import Binarizer, PolynomialFeatures, MaxAbsScaler

Poly = PolynomialFeatures(degree=3, include_bias=True, interaction_only=False)
X_Poly = Poly.fit_transform(X)


Bin = Binarizer(threshold = 1.0)
X_Bin = Bin.fit_transform(X)

MAX = MaxAbsScaler(copy=True)
X_MAX = MAX.fit_transform(X)
```

```python

# Loop over different values of k
neig = np.arange(1, 30)
kfold = 10
val_accuracy = {'Poly':[], 'MAX':[], 'Bin':[]}
bestKnn = None
bestAcc = 0.0
bestScaling = None
# Loop over different values of k
for i, k in enumerate(neig):
    knn = KNeighborsClassifier(n_neighbors=k)
    # validation accuracy
    s1 = np.mean(cross_val_score(knn, X_Poly, y, cv=kfold))
    val_accuracy['Poly'].append(s1)
    s2 = np.mean(cross_val_score(knn, X_MAX, y, cv=kfold))
    val_accuracy['MAX'].append(s2)
    s3 = np.mean(cross_val_score(knn, X_Bin, y, cv=kfold))
    val_accuracy['Bin'].append(s3)
    if s1 > bestAcc:
        bestAcc = s1
        bestKnn = knn
        bestScaling = 'Poly'
    elif s2 > bestAcc:
        bestAcc = s2
        bestKnn = knn
        bestScaling = 'MAX'
    elif s3 > bestAcc:
        bestAcc = s3
        bestKnn = knn
        bestScaling = 'Bin'
        
        
        
# Plot
plt.figure(figsize=[13,8])
plt.plot(neig, val_accuracy['Poly'], label = 'CV Accuracy with std')
plt.plot(neig, val_accuracy['MAX'], label = 'CV Accuracy with MAX')
plt.plot(neig, val_accuracy['Bin'], label = 'CV Accuracy with Bin')
plt.legend()
plt.title('k value VS Accuracy')
plt.xlabel('Number of Neighbors')
plt.ylabel('Accuracy')
plt.xticks(neig)
plt.show()

print('Best Accuracy with feature scaling:', bestAcc)
print('Best kNN classifier:', bestKnn)
print('Best scaling:', bestScaling)
```
