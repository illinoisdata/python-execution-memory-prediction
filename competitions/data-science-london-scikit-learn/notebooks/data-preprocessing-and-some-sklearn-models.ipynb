## Checking correlations

We dont need features that have linear correlations between each other.

```python
def draw_correlations(df):
    correlations = df.corr()
    sns.set(rc={'figure.figsize': (12, 12)})
    sns.heatmap(correlations)
    
draw_correlations(df_train_data)
```

```python
df_train_data = df_train_data.drop([4, 12], axis=1)
df_train_data.head()
```

```python
draw_correlations(df_train_data)
```

## Data normalization

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_train_data = pd.DataFrame(scaler.fit_transform(df_train_data))
df_train_data.describe()
```

# Models training

## Train/test split

```python
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(df_train_data, df_train_labels, train_size=0.75, shuffle=True)
```

## SVM

```python
from sklearn.svm import SVC
from sklearn.metrics import f1_score
svc = SVC()
svc.fit(X_train, y_train)
y_pred = svc.predict(X_val)
print(f1_score(y_pred, y_val))
```

## Linear SVM

```python
from sklearn.svm import LinearSVC
svc_lin = LinearSVC()
svc_lin.fit(X_train, y_train)
y_pred = svc_lin.predict(X_val)
print(f1_score(y_pred, y_val))
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
rnd_forest = RandomForestClassifier(n_estimators=100, criterion="entropy", max_depth = 10)
rnd_forest.fit(X_train, y_train)
y_pred = rnd_forest.predict(X_val)
print(f1_score(y_pred, y_val))
```
