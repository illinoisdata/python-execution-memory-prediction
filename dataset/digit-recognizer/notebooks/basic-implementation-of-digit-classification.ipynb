```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
train_dir=pd.read_csv("/kaggle/input/digit-recognizer/train.csv")
test_dir=pd.read_csv("/kaggle/input/digit-recognizer/test.csv")
```

```python
train_dir.shape, test_dir.shape
```

```python
train_labels=train_dir['label']
train_images=train_dir.drop(columns='label')
```

```python
print(type(train_images))
print(type(train_labels))
```

```python
x_train, x_val, y_train, y_val = train_test_split(train_images, train_labels, test_size=0.2, random_state=42)
```

```python
classifier=RandomForestClassifier(random_state=42)
```

```python
classifier.fit(x_train, y_train)
```

```python
y_pred=classifier.predict(x_val)
```

```python
round(accuracy_score(y_val, y_pred), 2)
```

```python
test_prediction=classifier.predict(test_dir)
```

```python
output = pd.DataFrame({
    'ImageId': range(1, len(test_prediction) + 1),
    'Label': test_prediction
})

output.to_csv('submission.csv', index=False)
```
