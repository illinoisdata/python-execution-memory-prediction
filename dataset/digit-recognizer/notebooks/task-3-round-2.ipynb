```python
train = pd.read_csv("/kaggle/input/digit-recognizer/train.csv")
test = pd.read_csv("/kaggle/input/digit-recognizer/test.csv")
```

```python
print(train.shape)
print(test.shape)
```

```python
train.head()
```

```python
X = train.iloc[:, 1:]
y = train.iloc[:, 0]
```

```python
import matplotlib.pyplot as plt

def print_image(row, df):
    val = df.iloc[row, :].values
    val = val.reshape(28,28).astype('uint8')
    plt.imshow(val)
```

```python
print_image(5000, X)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
forest = RandomForestClassifier()
```

```python
forest.fit(X_train, y_train)
```

```python
pred = forest.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print("Classification Report")

print(classification_report(y_test, pred))
```

```python
print("Confusion Matrix")
print(confusion_matrix(y_test, pred))
```

```python
perdiction = forest.predict(test)

prediction = pd.Series(perdiction, name = "Label")
```

```python
submission = pd.concat([pd.Series(range(1,28001), name = "ImageId"), prediction], axis = 1)
```

```python
submission.to_csv("submission.csv", index=False)
```
