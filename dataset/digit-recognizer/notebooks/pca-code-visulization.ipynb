### Using PCA:

```python
# Now we will use the pca and try to reduce the no of cols and getting the same accuracy
```

```python
# Step-1: Standardize the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Standardizing the data
```

```python
## Step-2: Find covariance matrix and eigendecomposition of it
## But we will do this thing using sklearn

# PCA 
from sklearn.decomposition import PCA
pca = PCA(n_components=200) # n_components means how many dimenssion you want in your converted data
```

```python
X_train_trf = pca.fit_transform(X_train)
X_test_trf = pca.transform(X_test)

# It will first the find the covariance matrix of the data and then it will perform the eigendecomposition of the covariance matrix
# then it will give the best eigenvalues for direction and then it will pick the eigenvectors according to the number
# we provide it earlier at the time of object creation 
# and then it will do projection of acutal data into new cordinate system and gives us the final data that we can use
# for the model training purpose.
```

```python
X_train_trf.shape
```

```python
## Now we are performing the model training

knn = KNeighborsClassifier()
```

```python
knn.fit(X_train_trf,y_train) # now we will give it the pca converted data
```

```python
y_pred = knn.predict(X_test_trf) # it will not take much time as we reduce the sparsity of the data
```

```python
accuracy_score(y_test,y_pred)*100
```

```python
# look at the accuracy we achieve in lower dimension that is nearly closed from the acutal acc, nd we save much no of computations as we reduced the no of dimensions
```
