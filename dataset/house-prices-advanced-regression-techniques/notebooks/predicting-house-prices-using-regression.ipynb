```python
df_train=pd.read_csv('/kaggle/input/house-prices-advanced-regression-techniques/train.csv')
df_test=pd.read_csv('/kaggle/input/house-prices-advanced-regression-techniques/test.csv')
```

```python
numerical_cols = df_train.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = df_train.select_dtypes(include=['object']).columns

print("Numerical columns:", len(numerical_cols))
print("Categorical columns:", len(categorical_cols))
```

```python
# correlation between numerical columns
plt.figure(figsize=(12, 8))
plt.imshow(df_train[numerical_cols].corr(), cmap='coolwarm', interpolation='nearest')
plt.colorbar(label='Correlation Coefficient')
plt.title("Correlation Matrix of Numerical Features", fontsize=14)
plt.xticks(range(len(df_train[numerical_cols].corr().columns)), df_train[numerical_cols].corr().columns, rotation=90)
plt.yticks(range(len(df_train[numerical_cols].corr().columns)), df_train[numerical_cols].corr().columns)
plt.tight_layout()
plt.show()
```

```python
# filling missing values with median in numerical columns

df_train[numerical_cols] = df_train[numerical_cols].apply(pd.to_numeric, errors='coerce')
df_train[numerical_cols] = df_train[numerical_cols].fillna(df_train[numerical_cols].median())
```

```python
# filling the categorical columns with mode

df_train[categorical_cols] = df_train[categorical_cols].fillna(df_train[categorical_cols].mode().iloc[0])
```

# Regression model implementation

```python
# Encoding categorical columns
X = pd.get_dummies(df_train.drop(columns=['SalePrice'], errors='ignore'), columns=categorical_cols, drop_first=True)
y = df_train['SalePrice']

# Aligning columns in both datasets so that there is no change in the column names for testing dataset
X, df_test = X.align(df_test, join='left', axis=1, fill_value=0)

# since, we got teh error regarding NA values in df_test, so it is necessary to fill NA values with 0
df_test = df_test.fillna(0)
```

## **1. Linear regression**

```python
# Train Linear Regression
model = LinearRegression()
model.fit(X, y)
```

```python
# target value 
y_pred = model.predict(df_test)
```

```python
# creating dataframe from the predicted target 
sample_submission = pd.DataFrame({"Id": df_test["Id"],"SalePrice": y_pred})
```

```python
# Save submission file
sample_submission.to_csv("sample_submission.csv", index=False)
```
