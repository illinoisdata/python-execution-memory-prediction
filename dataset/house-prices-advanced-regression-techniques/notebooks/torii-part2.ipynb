```python
# 数値型のカラムを抽出して、リストにする
# 1. すべての数値型 ('number') のカラムを抽出
numeric_df = train_df.select_dtypes(include='number')

# 2. 抽出されたデータフレームのカラム名を取得し、リストに変換
numeric_columns_list = numeric_df.columns.tolist()

# 3. リストの各要素をダブルクォーテーションで囲む
quoted_columns_list = [f'"{col}"' for col in numeric_columns_list]

# 4. リストをカンマとスペースで区切られた文字列に結合
output_string = ", ".join(quoted_columns_list)

print(output_string)
```

```python
# 1. すべての文字列型 ('object') のカラムを抽出
string_df = train_df.select_dtypes(include='object')

# 2. 抽出されたデータフレームのカラム名を取得し、リストに変換
string_columns_list = string_df.columns.tolist()

# 3. リストの各要素をダブルクォーテーションで囲む
quoted_columns_list = [f'"{col}"' for col in string_columns_list]

# 4. リストをカンマとスペースで区切られた文字列に結合
output_string = ", ".join(quoted_columns_list)

print(output_string)
```

```python
# Xに、数値型のカラムを全部入れて、欠損は0埋め
X = train_df[["MSSubClass", "LotFrontage", "LotArea", "OverallQual", "OverallCond", "YearBuilt", "YearRemodAdd", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2", "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF", "LowQualFinSF", "GrLivArea", "BsmtFullBath", "BsmtHalfBath", "FullBath", "HalfBath", "BedroomAbvGr", "KitchenAbvGr", "TotRmsAbvGrd", "Fireplaces", "GarageYrBlt", "GarageCars", "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch", "3SsnPorch", "ScreenPorch", "PoolArea", "MiscVal", "MoSold", "YrSold", "MSZoning", "Street", "Alley", "LotShape", "LandContour", "Utilities", "LotConfig", "LandSlope", "Neighborhood", "Condition1", "Condition2", "BldgType", "HouseStyle", "RoofStyle", "RoofMatl", "Exterior1st", "Exterior2nd", "MasVnrType", "ExterQual", "ExterCond", "Foundation", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2", "Heating", "HeatingQC", "CentralAir", "Electrical", "KitchenQual", "Functional", "FireplaceQu", "GarageType", "GarageFinish", "GarageQual", "GarageCond", "PavedDrive", "PoolQC", "Fence", "MiscFeature", "SaleType", "SaleCondition"]].fillna(0)

# ここで対数を取る
y = np.log1p(train_df["SalePrice"])

# Xと同様の処理
X_test = test_df[["MSSubClass", "LotFrontage", "LotArea", "OverallQual", "OverallCond", "YearBuilt", "YearRemodAdd", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2", "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF", "LowQualFinSF", "GrLivArea", "BsmtFullBath", "BsmtHalfBath", "FullBath", "HalfBath", "BedroomAbvGr", "KitchenAbvGr", "TotRmsAbvGrd", "Fireplaces", "GarageYrBlt", "GarageCars", "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch", "3SsnPorch", "ScreenPorch", "PoolArea", "MiscVal", "MoSold", "YrSold", "MSZoning", "Street", "Alley", "LotShape", "LandContour", "Utilities", "LotConfig", "LandSlope", "Neighborhood", "Condition1", "Condition2", "BldgType", "HouseStyle", "RoofStyle", "RoofMatl", "Exterior1st", "Exterior2nd", "MasVnrType", "ExterQual", "ExterCond", "Foundation", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2", "Heating", "HeatingQC", "CentralAir", "Electrical", "KitchenQual", "Functional", "FireplaceQu", "GarageType", "GarageFinish", "GarageQual", "GarageCond", "PavedDrive", "PoolQC", "Fence", "MiscFeature", "SaleType", "SaleCondition"]].fillna(0)
```

```python
X
```

```python
# ダミー変数を使って、文字列をbool値で認識できるようにする
X_2 = pd.get_dummies(X)
X_test_2 = pd.get_dummies(X_test)
```

```python
X_2
```

```python
from sklearn.model_selection import train_test_split
```

注：説明のわかりやすさのためにここに書いているが、↑のようなimport系は本当は一番上にまとめるのがお行儀の良い書き方

```python
# X_2を使う
X_train, X_valid, y_train, y_valid = train_test_split(X_2, y, random_state=0)
```

### モデルの構築

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

### モデルの適用と評価

```python
# X_test_2のカラムをX_2のカラム構造に揃えておく（理由：カラム構造が異なると、モデルが適用できないため）
#      X_2にあるがX_test_2にないカラムは、新しく追加され '0' で埋められます。
X_test_2 = X_test_2.reindex(columns=X_2.columns, fill_value=0)
```

```python
# X_train , X_validは、X_2が使われている
# X_test_2を使う
pred_train = model.predict(X_train)
pred_valid = model.predict(X_valid)
pred_test = model.predict(X_test_2)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
print(np.sqrt(mean_squared_error(y_train, pred_train)), np.sqrt(mean_squared_error(y_valid, pred_valid)))
```
