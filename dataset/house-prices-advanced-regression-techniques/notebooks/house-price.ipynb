```python
majpearson = []
for col in df:
    if df[col].dtype == np.float64 or df[col].dtype == np.int64:
        pearson = round(stats.pearsonr(df[col], df['SalePrice'])[0],2)
        if pearson > .5:
            majpearson.append(col)
```

```python
from sklearn.decomposition import PCA

pca_livarea = PCA(
    n_components=1,
)

df_livare = pd.DataFrame(
    pca_livarea.fit_transform(
        df_mjprs[[
            'TotalBsmtSF_scaled','1stFlrSF_scaled','GrLivArea_scaled'
        ]]
    ),
    columns=['PCA_livarea']
)
df_livare
```

```python
plt.figure(figsize=(6,4))

sns.histplot(
    data=df_livare,
    x='PCA_livarea',
    kde=True
)

plt.grid(
    linestyle='dotted',
    linewidth=.7
)
plt.tight_layout()
plt.show()
```

```python
pca_livarea.explained_variance_ratio_
```

```python
# FACTOR DE CAPACIDAD DEL GARAGE
pca_garage = PCA(
    n_components=1,
)

df_garage = pd.DataFrame(
    pca_garage.fit_transform(
        df_mjprs[[
            'GarageArea_scaled','GarageCars_scaled'
        ]]
    ),
    columns=['PCA_garage']
)
df_garage
```

```python
plt.figure(figsize=(6,4))

sns.histplot(
    data=df_garage,
    x='PCA_garage',
    kde=True
)

plt.grid(
    linestyle='dotted',
    linewidth=.7
)
plt.tight_layout()
plt.show()
```

```python
df_mjprs['isRemod'] = [
    True if df_mjprs.loc[index,'YearRemodAdd'] != df_mjprs.loc[index,'YearBuilt'] else False for index in df_mjprs.index
]
```

```python
df_final = pd.concat(
    [
        df_mjprs,df_livare,df_garage
    ],
    axis=1
)
df_final
```

```python
X = df_final.drop('SalePrice',axis=1)
y = df_final['SalePrice']
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
import numpy as np

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=int(np.random.normal(100,10)))

bosque = RandomForestRegressor()
bosque.fit(X_train,y_train)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

predict = bosque.predict(X_test)

mae = mean_absolute_error(y_test,predict)
rmse = mean_squared_error(y_test,predict)**.5
r2 = r2_score(y_test,predict)

print(f'''
MAE = {mae:.3f}
RMSE = {rmse:.3f}
R2 = {r2:.3f}
''')
```

```python
df_test = ''

for dir,_,files in os.walk('/kaggle/input'):
    for file in files:
        if file.startswith('test'):
            df_test = pd.DataFrame(
                pd.read_csv(
                    os.path.join(
                        dir,
                        file
                    )
                )
            )

df_test.head()
```

```python
df_test[majpearson[0:-1]]
```

```python
df_sub = df_test[majpearson[0:-1]]
```

```python
t = pd.DataFrame(
    df_sub.isna().sum(),
    columns=['column']
)
vnan = t.index[
    t['column'] > 0
]
```

```python
for c in vnan:
    df_sub[c] = df_sub[c].fillna(df_sub[c].mean())
```

```python
df_sub
```

```python
scaler_test = StandardScaler()

df_sub['TotalBsmtSF_scaled'] = scaler_test.fit_transform(df_sub[['TotalBsmtSF']].astype(float))
df_sub['1stFlrSF_scaled'] = scaler_test.fit_transform(df_sub[['1stFlrSF']].astype(float))
df_sub['GrLivArea_scaled'] = scaler_test.fit_transform(df_sub[['GrLivArea']].astype(float))

df_sub['GarageArea_scaled'] = scaler_test.fit_transform(df_sub[['GarageArea']].astype(float))
df_sub['GarageCars_scaled'] = scaler_test.fit_transform(df_sub[['GarageCars']].astype(float))
```

```python
df_sub
```

```python
pca_test_livarea = PCA(
    n_components=1,
)

df_test_livare = pd.DataFrame(
    pca_test_livarea.fit_transform(
        df_sub[[
            'TotalBsmtSF_scaled','1stFlrSF_scaled','GrLivArea_scaled'
        ]]
    ),
    columns=['PCA_livarea']
)
df_test_livare
```

```python
pca_test_garage = PCA(
    n_components=1,
)

df_test_garage = pd.DataFrame(
    pca_test_garage.fit_transform(
        df_sub[[
            'GarageArea_scaled','GarageCars_scaled'
        ]]
    ),
    columns=['PCA_garage']
)
df_test_garage
```

```python
df_sub['isRemod'] = [
    True if df_sub.loc[index,'YearBuilt'] != df_sub.loc[index,'YearRemodAdd'] else False for index in df_sub.index
]
```

```python
df_test_final = pd.concat(
    [
        df_sub,df_test_livare,df_test_garage
    ],
    axis=1
)
df_test_final
```

```python
test_predict = bosque.predict(df_test_final)
```

```python
df_predict = pd.concat(
    [
        df_test['Id'],
        pd.DataFrame(
            test_predict,
            columns=['SalePrice']
        )
    ],
    axis=1
)
```
