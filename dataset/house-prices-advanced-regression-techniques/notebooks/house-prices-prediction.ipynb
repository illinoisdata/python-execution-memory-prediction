```python
import plotly.express as px
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
```

```python
#Read csv file in Dataframe
test_df=pd.read_csv("/kaggle/input/house-prices-advanced-regression-techniques/test.csv")
#In predicting the price of house there is no involement of 'Id" columns so we drop it 
test_df=test_df.drop(columns="Id")
```

```python
test_df.head()
#Remove Object dtype feature columns
test_num_df=test_df.select_dtypes(include=["float64","int64"])
```

```python
#Read train.csv file into DataFrame
train_df=pd.read_csv("/kaggle/input/house-prices-advanced-regression-techniques/train.csv")
train_df=train_df.drop(columns="Id")
train_df.head()
```

```python
#list all the types of data from our dataset
list(set(train_df.dtypes))
```

# House Price Distribution plot

```python
print(train_df["SalePrice"].describe())
plt.figure(figsize=(10,5))
train_df["SalePrice"].hist()
```

```python
#From here we just only see numerical featrue columns 
df_num=train_df.select_dtypes(include=["float64","int64"])
df_num.head()
df_num.info()
```

```python
#Distribution of Numerical features 
df_num.hist(figsize=(16,20),bins=50);
```

```python
#Finding the null values in all the features columns 
pd.DataFrame(df_num.isnull().sum())
```

```python
#fill all the null values using the ffill method 
df_num["LotFrontage"].fillna(method="ffill",inplace=True)
df_num.isnull().sum()
```

```python
#In Linear Regression model we need split our train data into X and y
target="SalePrice"
X=df_num.drop(columns=target)
y=df_num[target]
```

```python
#instantiate the Linear Regression model
model=LinearRegression()
```

```python
model.fit(X,y)
```

```python
model.predict(test_num_df).head()
```

```python
#fill the NaN value of the test set so that we can feed to the model
X_test=test_num_df.fillna(method="ffill")
```

```python
model.predict(X_test)
```

```python
#mean_absolute_error of the test data 
MAE_train_data=mean_absolute_error(y,model.predict(X))
MAE
```
