```python
train_df = pd.read_csv("../input/house-prices-advanced-regression-techniques/train.csv")
test_df = pd.read_csv("../input/house-prices-advanced-regression-techniques/test.csv")
sample_df = pd.read_csv("../input/house-prices-advanced-regression-techniques/sample_submission.csv")
```

```python
X = train_df[["BsmtFinSF1","LotArea","BsmtFinSF2","GarageArea"]]
# ここで対数を取る
y = np.log1p(train_df["SalePrice"])
X_test = test_df[["BsmtFinSF1","LotArea","BsmtFinSF2","GarageArea"]]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X, y, random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
# 各カラムの欠損値数
print("=== 各カラムの欠損値数 ===")
missing_counts = X.isnull().sum()
print(missing_counts)
```

```python
# 各カラムの欠損値数
print("=== 各カラムの欠損値数 ===")
missing_counts = X_test.isnull().sum()
print(missing_counts)
```

```python
#欠損値を平均値いれる
X_test["GarageArea"] = X_test["GarageArea"].fillna(X_test["GarageArea"].mean())
```

```python
X_test["BsmtFinSF1"] = X_test["BsmtFinSF1"].fillna(X_test["BsmtFinSF1"].mean())
```

```python
X_test["BsmtFinSF2"] = X_test["BsmtFinSF2"].fillna(X_test["BsmtFinSF2"].mean())
```

```python
# 各カラムの欠損値数
print("=== 各カラムの欠損値数 ===")
missing_counts = X_test.isnull().sum()
print(missing_counts)
```

```python
model.fit(X_train, y_train)
```

```python
pred_train = model.predict(X_train)
pred_valid = model.predict(X_valid)
pred_test = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
print(mean_squared_error(y_train, pred_train), mean_squared_error(y_valid, pred_valid))
```

```python
sub_df = pd.DataFrame({
    "Id": sample_df["Id"], 
    "SalePrice": pred_test
})
```
