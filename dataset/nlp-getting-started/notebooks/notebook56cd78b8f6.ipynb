```python
import pandas as pd
import re
import string
import seaborn as sns
import matplotlib.pyplot as plt
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

def count_emojis(text):
    emoji_pattern = re.compile("[" 
                               u"\U0001F600-\U0001F64F"
                               u"\U0001F300-\U0001F5FF"
                               u"\U0001F680-\U0001F6FF"
                               u"\U0001F1E0-\U0001F1FF"
                               u"\U00002702-\U000027B0"
                               u"\U000024C2-\U0001F251"
                               "]+", flags=re.UNICODE)
    return len(emoji_pattern.findall(text))

def count_html(text):
    html_pattern = re.compile(r'<.*?>')
    return len(html_pattern.findall(text))

def count_urls(text):
    url_pattern = re.compile(r'https?://\S+|www\.\S+')
    return len(url_pattern.findall(text))

def count_punctuations(text):
    return sum([1 for c in text if c in string.punctuation])

def count_stopwords(text):
    STOPWORDS = set(stopwords.words("english"))
    words = text.lower().split()
    return sum([1 for w in words if w in STOPWORDS])

df_train = pd.read_csv("/kaggle/input/nlp-getting-started/train.csv")

df_train['emoji_count'] = df_train['text'].apply(count_emojis)
df_train['html_count'] = df_train['text'].apply(count_html)
df_train['url_count'] = df_train['text'].apply(count_urls)
df_train['punct_count'] = df_train['text'].apply(count_punctuations)
df_train['stopword_count'] = df_train['text'].apply(count_stopwords)

sns.set(style="whitegrid")
plt.figure(figsize=(15, 10))

features = ['emoji_count', 'html_count', 'url_count', 'punct_count', 'stopword_count']
titles = ['Emoji Count', 'HTML Tag Count', 'URL Count', 'Punctuation Count', 'Stopword Count']

for i, feature in enumerate(features):
    plt.subplot(2, 3, i + 1)
    sns.histplot(df_train[feature], bins=30, kde=False)
    plt.title(titles[i])
    plt.xlabel("Count")
    plt.ylabel("Frequency")

plt.tight_layout()
plt.suptitle("Distribution of Noise Elements in Training Data", fontsize=16, y=1.02)
plt.show()
```
