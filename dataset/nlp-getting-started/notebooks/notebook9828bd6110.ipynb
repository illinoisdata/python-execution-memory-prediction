```python
# 📦 Step 1: Install and import dependencies
from fastai.text.all import *
import pandas as pd

# 📂 Step 2: Load the data
path = Path('/kaggle/input/nlp-getting-started/')
train_df = pd.read_csv(path/'train.csv')
test_df = pd.read_csv(path/'test.csv')
sample_submission = pd.read_csv(path/'sample_submission.csv')

# 🧼 Step 3: Preprocessing (optional)
# Remove NaN texts (just in case)
train_df['text'] = train_df['text'].fillna('')
test_df['text'] = test_df['text'].fillna('')

# 🧠 Step 4: Prepare the DataLoaders
dls = TextDataLoaders.from_df(train_df, text_col='text', label_col='target', 
                              valid_pct=0.2, is_lm=False)

# 🏋️‍♀️ Step 5: Fine-tune a pretrained language model for classification
learn = text_classifier_learner(dls, AWD_LSTM, metrics=accuracy)
learn.fine_tune(4)

# 🔍 Step 6: Prepare test set
# You must prepare the test_dl using the test dataframe
test_dl = dls.test_dl(test_df)

# 🔮 Step 7: Get predictions
preds, _ = learn.get_preds(dl=test_dl)
final_preds = (preds[:, 1] > 0.5).int()  # Class 1 = disaster, 0 = not

# 📝 Step 8: Create submission file
submission = pd.DataFrame({'id': test_df['id'], 'target': final_preds})
submission.to_csv('submission.csv', index=False)

# ✅ Done! Now go to "Submit Predictions" and upload submission.csv
```
