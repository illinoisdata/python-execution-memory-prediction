```python
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import r2_score
from statistics import mode

train = pd.read_csv("../input/titanic/train.csv")
test  = pd.read_csv('../input/titanic/test.csv')
```

```python
X_train = train.drop(['Survived', 'PassengerId'], axis = 1)
y_train = train['Survived']
X_test = test.drop(['PassengerId'], axis = 1)
```

```python
X_train.info() # remove cabin 
```

```python
sns.countplot(x = X_train['Embarked'])
```

```python
X_train.drop(['Cabin'], axis = 1,inplace = True)
X_test.drop(['Cabin'], axis = 1,inplace = True)
```

```python
print(f"Train: ",X_train.isnull().sum())
print()
print(f"Test: ",X_test.isnull().sum())
# remove age as well
# fill Fare(with mean), Embarked(with "S"), 
```

```python
X_test.Fare = X_test.Fare.fillna(X_test.Fare.mean())
X_train['Embarked'] = X_train['Embarked'].fillna("S")
```

```python
print(f"Train: ",X_train.isnull().sum())
print()
print(f"Test: ",X_test.isnull().sum())
```

```python
# now lets do encoding and Feature Engineering stuff

ports = pd.get_dummies(X_train.Embarked , prefix='Embarked')
X_train = X_train.join(ports)
ports_test = pd.get_dummies(X_test.Embarked, prefix  = 'Embarked')
X_test = X_test.join(ports_test)

X_train.drop(['Embarked'], axis=1, inplace=True)
X_train.Sex = X_train.Sex.map({'male':0, 'female':1})
X_test.Sex = X_test.Sex.map({'male':0, 'female':1})

X_train.drop(["Name" ,'Ticket'], axis=1, inplace=True) 
X_test.drop(["Name" ,'Ticket'], axis=1, inplace=True)



X_train['TotalFamMembers'] = X_train['SibSp'] + X_train['Parch'] + 1
X_test['TotalFamMembers'] = X_test['SibSp'] + X_test['Parch'] + 1

X_train = X_train.drop(['SibSp', 'Parch'], axis = 1)
X_test = X_test.drop(['SibSp', 'Parch'], axis = 1)
```

```python
X_test.drop(['Embarked'], axis=1, inplace=True)
```

```python
X_train['Age'].info()
print(X_train['Age'].isnull().sum())
print(X_train['Age'].dtype)
```

```python
print(X_train['Age'].isnull().sum())
print(X_test['Age'].isnull().sum()) 
# above two indicates to remove age column
```

```python
X_train = X_train.drop(['Age'], axis = 1)
X_test = X_test.drop(['Age'], axis = 1)
```

# *Logistic Regression*

do train test split for accuracy

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_train, y_train, test_size=0.25, random_state=5)
```

```python
model_lgr = LogisticRegression(fit_intercept = True, penalty = 'l2', solver = 'lbfgs', C = 1.0, max_iter = 50000) # these all parameters are actually default
model_lgr.fit(X_train, y_train)
```

```python
y_test.shape
```

```python
X_test.shape
```

```python
predictions = model_lgr.predict(X_test)
acc = model_lgr.score(X_test, y_test)
print("Accuracy: ", acc)
```
