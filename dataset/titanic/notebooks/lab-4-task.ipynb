```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from keras import Sequential, Input
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.metrics import Precision, Recall
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
import shap
shap.initjs()
from lime.lime_tabular import LimeTabularExplainer
```

```python
df = pd.read_csv('/kaggle/input/titanic/train.csv') 
df['Age'] = df['Age'].fillna(df['Age'].median()) 
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0]) 
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1}) 
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True) 
df = df.drop(columns=['Cabin', 'Name', 'Ticket'])  
X = df[['Pclass', 'Sex', 'Age', 'Fare', 'SibSp', 'Parch','Embarked_S','Embarked_Q']] 
y = df['Survived'] 
```

```python
scaler = StandardScaler() 
X_scaled = scaler.fit_transform(X) 
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, 
test_size=0.2, random_state=42) 
```

```python
print("Train data shape:",X_train.shape, "Train Label shape", y_train.shape) 
print("Test data shape:",X_test.shape, "Test Label shape", y_test.shape)
```

```python
model = Sequential([ 
Input(shape=(X_train.shape[1],)), 
layers.Dense(16, activation='relu'), 
layers.Dense(8, activation='relu'), 
layers.Dense(1, activation='sigmoid')
]) 
model.summary()   
```

```python
model.compile(optimizer='adam', 
loss='binary_crossentropy', 
metrics=['accuracy', 
Precision(name='precision'), 
Recall(name='recall')]) 
```

```python
early_stop = EarlyStopping(monitor='val_accuracy', patience=3, restore_best_weights=True)
history = model.fit(
    X_train, y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1,
    callbacks=[early_stop]
)
```

```python
plt.figure(figsize=(10,6))
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.plot(history.history['precision'], label='Train Precision')
plt.plot(history.history['val_precision'], label='Val Precision')
plt.plot(history.history['recall'], label='Train Recall')
plt.plot(history.history['val_recall'], label='Val Recall')
plt.xlabel('Epoch')
plt.ylabel('Score')
plt.title('Training Progress (Accuracy, Precision, Recall)')
plt.legend()
plt.show()
```

```python
test_loss, test_accuracy, test_precision, test_recall = model.evaluate(X_test, y_test, verbose=0)
print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print("Test Precision:", test_precision)
print("Test Recall:", test_recall)
```

```python
X_train_np = X_train.values
X_test_np = X_test.values
```

```python
lime_explainer = LimeTabularExplainer(
    X_train_np,
    mode='classification',
    training_labels=y_train,
    feature_names=X.columns.tolist(),
    discretize_continuous=True,
    random_state=42
)
```

```python
def lime_predict_fn(x):
    preds = model.predict(x)
    return np.hstack((1 - preds, preds))

j = 1
exp = lime_explainer.explain_instance(
    X_test_np[j],
    lime_predict_fn,
    num_features=6
)
```
