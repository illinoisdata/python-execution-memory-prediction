```python
import os

path = "/kaggle/input/titanic"
train = os.path.join(path, "train.csv")
test = os.path.join(path, "test.csv")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder

```

```python
train_data = pd.read_csv(train)
test_data = pd.read_csv(test)
```

```python
train_data.head()
```

```python
import matplotlib.pyplot as plt


sex_counts = train_data['Sex'].value_counts()

plt.pie(sex_counts, labels=sex_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Sex')
plt.show()
```

```python
train_data.info()
```

```python
# Compute correlation matrix (numeric only)
corr_matrix = train_data.corr(numeric_only=True)

# Sort correlations with 'Survived' in descending order
corr_with_survived = corr_matrix['Survived'].sort_values(ascending=False)

print(corr_with_survived)
```

```python
le = LabelEncoder()

# Encode 'Sex'
train_data['Sex'] = le.fit_transform(train_data['Sex'])
test_data['Sex'] = le.transform(test_data['Sex'])

# Fill missing values
for col in ['Age', 'Fare']:
    train_data[col].fillna(train_data[col].median(), inplace=True)
    test_data[col].fillna(test_data[col].median(), inplace=True)

# Fill missing 'Embarked' with most frequent value
train_data['Embarked'].fillna(train_data['Embarked'].mode()[0], inplace=True)
test_data['Embarked'].fillna(test_data['Embarked'].mode()[0], inplace=True)

# Encode 'Embarked'
train_data['Embarked'] = le.fit_transform(train_data['Embarked'])
test_data['Embarked'] = le.transform(test_data['Embarked'])

# ===== Feature Selection =====
features = ['Age', 'SibSp', 'Parch', 'Fare']
X = train_data[features]
y = train_data['Survived']
X_test = test_data[features]

# ===== Model Training =====
model = LogisticRegression(max_iter=1000)
model.fit(X, y)

# ===== Prediction =====
predictions = model.predict(X_test)

# ===== Create Submission File =====
submission = pd.DataFrame({
    'PassengerId': test_data['PassengerId'],
    'Survived': predictions
})

submission.to_csv('/kaggle/working/submission.csv', index=False)

print("✅ submission.csv file saved successfully in /kaggle/working/")
```
