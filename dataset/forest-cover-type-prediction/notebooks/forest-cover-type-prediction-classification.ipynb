> **Since 'Slope' column has outliers, we apply sqrt transformation to it to reduce skewness, thereby reducing effect of outliers**

```python
#BEFORE
print(train['Slope'].skew())
```

```python
#AFTER
#min_value = abs(df_train['Slope'].min())
print(np.sqrt(train['Slope']+1).skew())#can also be min_value instead of 1, but result is close to 0 with usage of 1
```

```python
#APPLYING THE TRANSFORMATION TO ORIGINAL TRAIN AND TEST
df_train['Slope']=np.sqrt(df_train['Slope']+1)
df_test['Slope']=np.sqrt(df_test['Slope']+1)
```

***HD_Hydrology***

> **Since we have both HD_Hydrology and VD_Hydrology in our data<br/>
we can apply a transformation c2=a2+b2, thus creating new feature 'c' - 'D_Hydrology'**

```python
#transforming original df train
df_train['D_Hydrology'] = np.sqrt(df_train['HD_Hydrology']**2 + df_train['VD_Hydrology']**2)
```

```python
#transforming original df test
df_test['D_Hydrology'] = np.sqrt(df_test['HD_Hydrology']**2 + df_test['VD_Hydrology']**2)
```

```python
train.drop(['HD_Hydrology', 'VD_Hydrology'], axis=1,inplace=True)
```

# Models

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
sd=StandardScaler()
```

```python
df_train.iloc[:, :8] = sd.fit_transform(df_train.iloc[:, :8])
df_test.iloc[:,:8] = sd.transform(df_test.iloc[:, :8])
```

```python
#now scaling D_Hydrology
sd1=StandardScaler()
df_train.iloc[:, -2:-1] = sd1.fit_transform(df_train.iloc[:, -2:-1])
df_test.iloc[:,-1:] = sd1.transform(df_test.iloc[:, -1:])
```

```python
X = df_train.iloc[:, :-2].copy()
X['D_Hydrology'] = df_train.iloc[:, -2:-1].copy()
y = df_train.iloc[:,-1:].copy()
```

```python
X_test = df_test.iloc[:, :-1].copy()
X_test['D_Hydrology'] = df_test.iloc[:, -1:].copy()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_valid,y_train,y_valid=train_test_split(X,y,test_size=0.20)
```

```python
X = np.array(X)
y = np.array(y).reshape(-1)
X_test = np.array(X_test)
```

```python
X_train = np.array(X_train)
X_valid = np.array(X_valid)
y_train = np.array(y_train).reshape(-1)
y_valid = np.array(y_valid).reshape(-1)
```

***Logistic Regression***

```python
from sklearn.linear_model import LogisticRegression
log_reg=LogisticRegression(max_iter=500)
log_reg.fit(X_train,y_train)
```

```python
pred=log_reg.predict(X_valid)
print(accuracy_score(y_valid,pred))
```

***KNN Algorithm***

```python
l=[i for i in range(1,11)]
accuracy=[]
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
for i in l:
    knn=KNeighborsClassifier(n_neighbors=i,weights='distance')
    knn.fit(X_train,y_train)
    pred=knn.predict(X_valid)
    accuracy.append(accuracy_score(y_valid,pred))
print(max(accuracy))
```

```python
plt.plot(l,accuracy)
plt.title('knn_accuracy plot')
plt.xlabel('neighbors')
plt.ylabel('accuracy')
plt.grid()
```

***SVM's***

```python
# Support Vector Machines
from sklearn.svm import SVC
svm=SVC(kernel='rbf')
svm.fit(X_train,y_train)
pred=(svm.predict(X_valid))
print(accuracy_score(y_valid,pred))
```

***Random Forest Classifier***

```python
from sklearn.ensemble import RandomForestClassifier
rand=RandomForestClassifier()
rand.fit(X_train,y_train)
pred=rand.predict(X_valid)
print(accuracy_score(y_valid,pred))
```

***Naive Bayes***

```python
from sklearn.naive_bayes import GaussianNB
nb=GaussianNB()
nb.fit(X_train,y_train)
pred=nb.predict(X_valid)
print(accuracy_score(y_valid,pred))
```
