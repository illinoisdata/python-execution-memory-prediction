```python
lab = [
        'Spruce/Fir', 
        'Lodgepole Pine', 
        'Ponderosa Pine', 
        'Cottonwood/Willow', 
        'Aspen', 
        'Douglas-fir', 
        'Krummholz'
]
plt.figure(figsize=(7,4))
sns.countplot(Xtrain, x='Cover_Type')
plt.xticks(ticks=[0,1,2,3,4,5,6], labels =lab, rotation=90)
plt.title('Class Distribution')
plt.show()
```

```python
X = Xtrain.drop(['Id', 'Cover_Type'], axis=1)
scaled_x = StandardScaler().fit_transform(X)
```

```python
pca = PCA(n_components=2)
data_pca = pca.fit_transform(scaled_x)
data_pca_df = pd.DataFrame(data=data_pca, columns=['pca1', 'pca2'])
```

```python
plt.figure(figsize=(10,10))
p = sns.scatterplot(data=data_pca_df, x='pca1', y='pca2', hue=Xtrain.Cover_Type, palette='coolwarm', s=12)
p.legend(handles=p.legend_.legend_handles, labels=lab, \
         bbox_to_anchor=(1.04, 0), loc="lower left", borderaxespad=0)
plt.show()
```

```python
val_per = []
```

```python
X_train, X_val, y_train, y_val = train_test_split(X, Xtrain.Cover_Type, test_size=0.2, random_state=42)
```

## Decision Tree

```python
dc_tree = DecisionTreeClassifier()
dc_tree.fit(X_train, y_train)
train_pred = dc_tree.predict(X_train)
train_acc = accuracy_score(y_train, train_pred)

val_pred = dc_tree.predict(X_val)
val_acc = accuracy_score(y_val, val_pred)
print(train_acc, val_acc)

add_inp = {'name':'dc_tree','accuracy':val_acc, 'model':dc_tree}

val_per.append(add_inp)
```

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

```python
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train, y_train)

train_pred = log_reg.predict(X_train)
train_acc = accuracy_score(y_train, train_pred)

val_pred = log_reg.predict(X_val)
val_acc = accuracy_score(y_val, val_pred)
print(train_acc, val_acc)

add_inp = {'name':'log_reg','accuracy':val_acc, 'model':log_reg}

val_per.append(add_inp)
```

```python
NB = GaussianNB()
NB.fit(X_train, y_train)

train_pred = NB.predict(X_train)
train_acc = accuracy_score(y_train, train_pred)

val_pred = NB.predict(X_val)
val_acc = accuracy_score(y_val, val_pred)
print(train_acc, val_acc)

add_inp = {'name':'NB','accuracy':val_acc, 'model':NB}

val_per.append(add_inp)
```

```python
KNN = KNeighborsClassifier()
KNN.fit(X_train, y_train)

train_pred = KNN.predict(X_train)
train_acc = accuracy_score(y_train, train_pred)

val_pred = KNN.predict(X_val)
val_acc = accuracy_score(y_val, val_pred)
print(train_acc, val_acc)

add_inp = {'name':'KNN','accuracy':val_acc, 'model':KNN}

val_per.append(add_inp)
```

```python
SVM = SVC()
SVM.fit(X_train, y_train)

train_pred = SVM.predict(X_train)
train_acc = accuracy_score(y_train, train_pred)

val_pred = SVM.predict(X_val)
val_acc = accuracy_score(y_val, val_pred)
print(train_acc, val_acc)

add_inp = {'name':'SVM','accuracy':val_acc, 'model':SVM}

val_per.append(add_inp)
```

```python
val_per
```

```python
model_performance = pd.DataFrame(val_per)
model_performance.drop_duplicates(subset='name', inplace=True)
best_model = model_performance.loc[model_performance['accuracy'].idxmax(), 'name']
colors = ['r' if i == best_model else 'b' for i in model_performance.name]
print(colors)
plt.figure(figsize=(6,5))
sns.barplot(model_performance, x='name', y='accuracy', palette=colors)
plt.axhline(y=model_performance.accuracy.max(), color='r', linestyle='--')
plt.show()
```
