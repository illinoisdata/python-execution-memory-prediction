```python
import pandas as pd

# Read the CSV file into a DataFrame named 'forrest'
train = pd.read_csv('/kaggle/input/forest-cover-type-prediction/train.csv')

# Display the first few rows of the DataFrame to verify the data has been loaded
print(train.head())
```

```python
train.info()
```

```python
#check if there is any class imbalancing
train['Cover_Type'].value_counts(normalize= True)
```

```python
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from keras.utils import to_categorical
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt

scaler = StandardScaler()

# Extract features and target variable
X = train.drop(['Id', 'Cover_Type'], axis=1)
y = train['Cover_Type']
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV

# Number of folds for cross-validation
num_folds = 5
kf = StratifiedKFold(n_splits=num_folds, shuffle=True, random_state=42)

# Initialize lists to store the results
accuracy_per_fold_svm = []

# Iterate through each fold
for fold, (train_index, val_index) in enumerate(kf.split(X, y)):
    print(f"Training SVM on Fold {fold + 1}...")

    # Split the data into training and validation sets
    X_fold_train, X_fold_val = X.iloc[train_index], X.iloc[val_index]
    y_fold_train, y_fold_val = y.iloc[train_index], y.iloc[val_index]

    # Scale the data using the previously fitted scaler
    X_fold_train_scaled = scaler.fit_transform(X_fold_train)
    X_fold_val_scaled = scaler.transform(X_fold_val)

    # Create and train the SVM model
    svm_model = SVC(kernel='rbf', C=1.0, gamma='scale')  # You can adjust kernel, C, and gamma parameters
    svm_model.fit(X_fold_train_scaled, y_fold_train)

    # Make predictions on the validation set
    y_pred = svm_model.predict(X_fold_val_scaled)

    # Evaluate accuracy for this fold
    accuracy = accuracy_score(y_fold_val, y_pred)
    accuracy_per_fold_svm.append(accuracy)

# Print the average performance across all folds
print("\nAverage Validation Accuracy (SVM): ", np.mean(accuracy_per_fold_svm))
```

```python
# Plot training & validation accuracy values for the last fold
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(accuracy_per_fold_svm, label='Validation Accuracy')
plt.title('Model accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```
