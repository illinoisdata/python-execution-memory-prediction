### Forest cover Prediction

In the Forest Cover Type competition, we were asked to predict the forest cover type (the predominant kind of tree cover) from cartographic variables.

We begun working as a group for the first two weeks and we then cjose to work individually thereafter.

At the beginning I we fitted the logistic regression model, decision tree model and the random forest model. This time we got a score of approximately 0.62.

The following we added the random forest model nd got a score of approximately 0.7.


From the fourth week I began working solo. 

I conducted extensive exploratory data analysis which gave me a lot insight into the dataset. This led me to performing feature engineering especially for some of the numerical variables. With the new enginnered features and other existing features I fitted the models we had fitted before, this time I also performed hyperparameter tuning. These are:
+ Logistic Regression
+ Decision Tree
+ Random forest

This time I also added the k_nearest neighbors model and the Extremely Randomized Trees (or Extra-Trees). The optimal model was the extra tree model 
and I got a competition score of 0.80795.
