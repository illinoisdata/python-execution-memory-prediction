# Project: Forest Cover 

## Introduction
The Forest Cover Type Prediction project is an exploratory and predictive modeling task aimed at determining the predominant tree cover in areas of the Roosevelt National Forest located in Northern Colorado. This endeavor is crucial for environmental conservation, resource management, and understanding ecological dynamics within these forested areas.

## Project Synopsis: 
The goal of this project is to construct a classification model capable of predicting the type of forest cover based on cartographic variables. 

## Dataset Overview:
The dataset utilized for this project is a comprehensive collection of cartographic variables measured from 30m x 30m patches of land. It encompasses a wide range of environmental attributes, including:

In the dataset for the Forest Cover Type prediction, the variables can be broadly categorized into two types: environmental features and categorical identifiers.

The target variable, Cover_Type, contains integer values 1 through 7. Each of these values corresponds to one of the 7 forest cover types.


### Environmental Features:

- Elevation (meters): The height above sea level at which the 30m x 30m patch of forest is located. Elevation can significantly influence the type of forest cover due to its impact on climate and soil conditions.

- Aspect (degrees azimuth): The compass direction that a slope faces. Aspect can affect the amount of sunlight and moisture a patch receives and hence influence the vegetation type.

- Slope (degrees): The steepness or incline of the terrain. Slope can affect soil erosion and water runoff, thus impacting the vegetation.

- Horizontal Distance to Hydrology (meters): The shortest horizontal distance from the center of the patch to the nearest surface water—streams, lakes, etc.

- Vertical Distance to Hydrology (meters): The vertical drop or rise from the patch to the nearest surface water. It is an indicator of potential water flow paths and soil moisture.

- Horizontal Distance to Roadways (meters): The shortest horizontal distance from the patch to the nearest roadway. Proximity to roadways can indicate human influence and potential access for forest management.

- Horizontal Distance to Fire Points (meters): The shortest horizontal distance from the patch to the nearest historical wildfire ignition points. This factor can be significant for understanding fire ecology and forest recovery processes.

- Hillshade Indexes (9am, Noon, 3pm): These variables measure the shade cast by topography at different times of day during the summer solstice. They are given on a scale of 0 to 255, where 0 is completely shaded and 255 is fully exposed to the sun.

## Categorical Identifiers:

- Wilderness Area (4 binary columns): These indicate whether the patch belongs to one of four designated wilderness areas within the forest. Each wilderness area has unique legal and ecological characteristics that may affect the cover type.

- Soil Type (40 binary columns): These represent the presence or absence of 40 different soil types, identified by the USFS soil taxonomy. Soil type can greatly influence the types of vegetation that a patch can support.

- Cover Type (7 categories): The target variable, it denotes the predominant kind of trees in the patch. The seven types are: Spruce/Fir, Lodgepole Pine, Ponderosa Pine, Cottonwood/Willow, Aspen, Douglas-fir, and Krummholz.

## Feature engineering

From our previous models fitted, we used the features as they are in the dataset. We will now consider engineering some new features by making some modifications on the existing features.

For instance we shall consider features such as Elevation + Vertical_Distance_To_Hydrology (which is the elevation from sea level of the nearest surface water feature from the actual observation site.

```python
forest_df['Total_Distance_To_Hydrology'] = (forest_df['Vertical_Distance_To_Hydrology']**2 + forest_df['Horizontal_Distance_To_Hydrology']**2)**0.5

forest_df['Elevation_Plus_Vertical_Hydrology'] = forest_df['Elevation'] + forest_df['Vertical_Distance_To_Hydrology']
forest_df['Elevation_Minus_Vertical_Hydrology'] = forest_df['Elevation'] - forest_df['Vertical_Distance_To_Hydrology']

forest_df['Hydrology_Plus_Fire_Points'] = forest_df['Horizontal_Distance_To_Hydrology'] + forest_df['Horizontal_Distance_To_Fire_Points']
forest_df['Hydrology_Minus_Fire_Points'] = forest_df['Horizontal_Distance_To_Hydrology'] - forest_df['Horizontal_Distance_To_Fire_Points']

forest_df['Hydrology_Plus_Roadways'] = forest_df['Horizontal_Distance_To_Hydrology'] + forest_df['Horizontal_Distance_To_Roadways']
forest_df['Hydrology_Minus_Roadways'] = forest_df['Horizontal_Distance_To_Hydrology'] - forest_df['Horizontal_Distance_To_Roadways']

forest_df['Fire_Points_Plus_Roadways'] = forest_df['Horizontal_Distance_To_Fire_Points'] + forest_df['Horizontal_Distance_To_Roadways']
forest_df['Fire_Points_Minus_Roadways'] = forest_df['Horizontal_Distance_To_Fire_Points'] - forest_df['Horizontal_Distance_To_Roadways']
```

## Model fitting

Prior to measuring the different metrix=cs of our models, we will design the functions and methods that we will use.

```python
def get_metrics(model, X_test, y_test):
    
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average='weighted')
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')

    return accuracy, precision, recall, f1


def predict(model, X_test, y_test):
    y_pred = model.predict(X_test)
    print('Accuracy:', accuracy_score(y_test, y_pred))
    print("F1 score (weighted average): ", f1_score(y_test, y_pred, average='weighted'))
    print("Precision (weighted average): ", precision_score(y_test, y_pred, average='weighted'))
    print("Recall (weighted average): ", recall_score(y_test, y_pred, average='weighted'))
    return y_pred


def plot_confusion_matrix(model, X_test, y_test, title):
    
    y_pred = model.predict(X_test)
    
    plt.figure(figsize=(7, 6))
    sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap="YlGnBu")
    plt.xlabel('Predicted Class')
    plt.ylabel('Actual Class')
    plt.title(title)
    plt.xticks(ticks=np.array(range(7))+0.5, labels=['1', '2', '3', '4', '5', '6', '7'])
    plt.yticks(ticks=np.array(range(7))+0.5, labels=['1', '2', '3', '4', '5', '6', '7'])
    plt.show()
    

def plot_roc_curves(model, X_test, y_test, title):
    
    plt.figure(figsize=(8, 3.5))

    for i in range(7):

        cover = i + 1

        y_test_proba = model.predict_proba(X_test)
        auc_test = roc_auc_score(y_test==cover, y_test_proba[:,i], multi_class='multinomial')
        y_test_fpr, y_test_tpr, _ = roc_curve(y_test==cover, y_test_proba[:,i])
        
        plt.plot(y_test_fpr, y_test_tpr, marker=None, linewidth=1.5, label=f'Cover Type {cover}, AUC: {np.round(auc_test, 4)}')

        plt.ylabel('True Positive Rate (Recall)', fontsize=12)
        plt.xlabel('False Positive Rate', fontsize=12)

        plt.legend(frameon=True, edgecolor='k', fontsize=9)

        plt.title(title)

    plt.show()
```
